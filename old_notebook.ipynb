```python
import pandas as pd



INPUT_PATH = "/content/drive/MyDrive/thesis data set/dataset1_combined.csv"
OUTPUT_PATH = "dataset1_10servers.csv"

N_LOW = 3
N_MED = 4
N_HIGH = 3

RANDOM_SEED = 42
MIN_COMPLETENESS = 0.90


#
# 1. COLUMNS THAT MUST BE PRESERVED

columns_to_keep = [
    "Series",
    "Time",
    "CPU_Load",
    "Power",
    "Network_RX",
    "Network_TX",
    "Inlet_Temperature",
    "CPU1_Temperature",
    "CPU2_Temperature",
    "Fan_speed1",
    "Fan_speed2",
    "Fan_speed3",
    "Fan_speed4",
    "Ram",
    "Ram_Used",
    "No_Of_Running_vms",
    "CPU_cores",
    "CPU_cores_used"
]

numeric_cols = [
    "CPU_Load",
    "Power",
    "Network_RX",
    "Network_TX",
    "Inlet_Temperature",
    "CPU1_Temperature",
    "CPU2_Temperature",
    "Fan_speed1",
    "Fan_speed2",
    "Fan_speed3",
    "Fan_speed4",
    "Ram",
    "Ram_Used",
    "No_Of_Running_vms",
    "CPU_cores",
    "CPU_cores_used"
]



# 2. LOAD ORIGINAL DATA


df = pd.read_csv(INPUT_PATH)

print("=" * 70)
print("ORIGINAL DATASET")
print("=" * 70)

print(f"Rows: {len(df):,}")
print(f"Columns: {len(df.columns)}")
print(f"Servers: {df['Series'].nunique()}")



# 3. CHECK THAT ALL 18 COLUMNS EXIST


missing_cols = [
    col for col in columns_to_keep
    if col not in df.columns
]

if missing_cols:
    raise ValueError(
        f"These columns are missing from the source file: {missing_cols}"
    )

print("\nAll 18 required columns are present.")



# 4. KEEP EXACTLY THE 18 ORIGINAL COLUMNS


df = df[columns_to_keep].copy()



# 5. CLEAN TIME
#

df["Time"] = pd.to_datetime(
    df["Time"],
    errors="coerce"
)



# 6. CONVERT MEASUREMENT COLUMNS TO NUMERIC


for col in numeric_cols:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )



# 7. CHECK MISSING VALUES


print("\nMissing values in original dataset:")

missing_original = df.isna().sum()

print(missing_original.to_string())



# 8. COUNT ROWS PER SERVER


server_counts = (
    df.groupby("Series")
    .size()
    .reset_index(name="n_rows")
)

expected_rows = (
    server_counts["n_rows"]
    .mode()
    .iloc[0]
)

server_counts["completeness"] = (
    server_counts["n_rows"]
    / expected_rows
)


print(
    f"\nExpected rows per server: "
    f"{expected_rows:,}"
)



# 9. KEEP SERVERS WITH >= 90% OF EXPECTED ROWS


eligible_servers = server_counts[
    server_counts["completeness"]
    >= MIN_COMPLETENESS
].copy()


print(
    f"Total servers: "
    f"{len(server_counts)}"
)

print(
    f"Servers passing >=90% completeness: "
    f"{len(eligible_servers)}"
)



# 10. CALCULATE AVERAGE CPU LOAD


eligible_ids = (
    eligible_servers["Series"]
    .tolist()
)

eligible_data = df[
    df["Series"].isin(eligible_ids)
].copy()


cpu_summary = (
    eligible_data
    .groupby("Series")
    .agg(
        mean_cpu_load=("CPU_Load", "mean"),
        max_cpu_load=("CPU_Load", "max"),
        mean_power=("Power", "mean"),
        mean_cpu_temp=("CPU1_Temperature", "mean")
    )
    .reset_index()
)


# Remove server only if average CPU cannot be calculated
cpu_summary = cpu_summary.dropna(
    subset=["mean_cpu_load"]
)



# 11. DIVIDE SERVERS INTO LOW / MEDIUM / HIGH CPU LOAD


cpu_summary["tier"] = pd.qcut(
    cpu_summary["mean_cpu_load"],
    q=3,
    labels=[
        "low",
        "medium",
        "high"
    ]
)



# 12. RANDOMLY SELECT 10 SERVERS


selected = []

selection_plan = [
    ("low", N_LOW),
    ("medium", N_MED),
    ("high", N_HIGH)
]


for tier, n_needed in selection_plan:

    candidates = cpu_summary[
        cpu_summary["tier"] == tier
    ].copy()

    print(
        f"\n{tier.upper()} tier: "
        f"{len(candidates)} available"
    )

    if len(candidates) < n_needed:

        raise ValueError(
            f"Not enough {tier} servers. "
            f"Need {n_needed}, "
            f"but only {len(candidates)} are available."
        )

    chosen = candidates.sample(
        n=n_needed,
        random_state=RANDOM_SEED
    )

    selected.append(chosen)


selected_servers = pd.concat(
    selected,
    ignore_index=True
)


selected_servers = selected_servers.sort_values(
    "mean_cpu_load"
)



# 13. DISPLAY SELECTED SERVERS


print("\n")
print("=" * 70)
print("SELECTED 10 SERVERS")
print("=" * 70)

print(
    selected_servers[
        [
            "Series",
            "tier",
            "mean_cpu_load",
            "max_cpu_load",
            "mean_power",
            "mean_cpu_temp"
        ]
    ].to_string(index=False)
)



# 14. GET IDs OF SELECTED SERVERS


selected_ids = (
    selected_servers["Series"]
    .tolist()
)



# 15. GO BACK TO ORIGINAL DATA
#
# IMPORTANT:
# We are NOT using the summary table here.
# We return to df and copy EVERY ORIGINAL ROW
# belonging to the selected 10 servers.


final_df = df[
    df["Series"].isin(selected_ids)
].copy()



# 16. KEEP EXACTLY THE ORIGINAL 18 COLUMNS


final_df = final_df[
    columns_to_keep
]



# 17. SORT BY SERVER AND TIME


final_df = (
    final_df
    .sort_values(
        ["Series", "Time"]
    )
    .reset_index(drop=True)
)



# 18. CHECK NUMBER OF ROWS FOR EACH SELECTED SERVER


print("\n")
print("=" * 70)
print("ROWS PER SELECTED SERVER")
print("=" * 70)

rows_selected = (
    final_df
    .groupby("Series")
    .size()
)

print(
    rows_selected.to_string()
)



# 19. CHECK WHETHER VALUES ACTUALLY CHANGE
# counts how many DIFFERENT values each column has
# for each selected server.
# 1 = constant for that server
# >1 = changes over time


print("\n")
print("=" * 70)
print("NUMBER OF UNIQUE VALUES PER SERVER")
print("=" * 70)

variation_check = (
    final_df
    .groupby("Series")[numeric_cols]
    .nunique(dropna=True)
)

print(
    variation_check.to_string()
)



# 20. SPECIAL CHECK FOR THE COLUMNS NOTICED


check_cols = [
    "Ram",
    "Ram_Used",
    "No_Of_Running_vms",
    "CPU_cores",
    "CPU_cores_used"
]

print("\n")
print("=" * 70)
print("CHECKING END COLUMNS")
print("=" * 70)

end_column_check = (
    final_df
    .groupby("Series")[check_cols]
    .nunique(dropna=True)
)

print(
    end_column_check.to_string()
)



# 21. CHECK MISSING VALUES IN FINAL DATASET


print("\n")
print("=" * 70)
print("MISSING VALUES IN FINAL DATASET")
print("=" * 70)

final_missing = (
    final_df
    .isna()
    .sum()
)

print(
    final_missing.to_string()
)



# 22. DISPLAY FIRST 20 ROWS


print("\n")
print("=" * 70)
print("FIRST 20 ROWS OF FINAL DATASET")
print("=" * 70)

print(
    final_df
    .head(20)
    .to_string(index=False)
)



# 23. SAVE


final_df.to_csv(
    OUTPUT_PATH,
    index=False,
    date_format="%Y-%m-%d %H:%M:%S"
)



# 24. FINAL CONFIRMATION


print("\n")
print("=" * 70)
print("FINAL DATASET SAVED")
print("=" * 70)

print(
    f"File: {OUTPUT_PATH}"
)

print(
    f"Servers: "
    f"{final_df['Series'].nunique()}"
)

print(
    f"Rows: "
    f"{len(final_df):,}"
)

print(
    f"Columns: "
    f"{len(final_df.columns)}"
)

print("\nFinal column order:")

for i, col in enumerate(
    final_df.columns,
    start=1
):
    print(f"{i}. {col}")
```

```python
import pandas as pd
import numpy as np
import os



BITBRAINS_DIR = "/content/drive/MyDrive/thesis data set bitbrain"

# Full filtered time-series output
OUTPUT_PATH = "/content/bitbrains_filtered_full.csv"

# Separate diagnostic summary
SUMMARY_PATH = "/content/bitbrains_filter_summary.csv"

MIN_COMPLETENESS = 0.90


#
# 1. FIND ALL AVAILABLE BITBRAINS FILES
#

files = []

for i in range(1, 101):

    path = f"{BITBRAINS_DIR}/{i}.csv"

    if os.path.exists(path):
        files.append(path)


print("=" * 80)
print("BITBRAINS FILES")
print("=" * 80)

print(f"Found {len(files)} VM files")


if len(files) == 0:

    raise FileNotFoundError(
        "No Bitbrains CSV files found in /content."
    )


#
# 2. LOAD EVERY BITBRAINS VM
#

all_vms = []


for path in files:

    vm_id = int(
        os.path.splitext(
            os.path.basename(path)
        )[0]
    )


    # Bitbrains uses semicolon delimiter
    vm = pd.read_csv(
        path,
        sep=";"
    )


    # Remove hidden spaces/tabs from column names
    vm.columns = [
        col.strip()
        for col in vm.columns
    ]


    # Add VM ID but otherwise preserve the raw structure
    vm.insert(
        0,
        "VM_ID",
        vm_id
    )


    all_vms.append(vm)


#
# 3. COMBINE THE RAW FILES
#

df = pd.concat(
    all_vms,
    ignore_index=True
)


print(
    f"Total raw observations: "
    f"{len(df):,}"
)

print(
    f"Total VMs: "
    f"{df['VM_ID'].nunique()}"
)



# 4. IDENTIFY ORIGINAL BITBRAINS COLUMN NAMES


timestamp_col = "Timestamp [ms]"
cpu_cores_col = "CPU cores"
cpu_capacity_col = "CPU capacity provisioned [MHZ]"
cpu_usage_mhz_col = "CPU usage [MHZ]"
cpu_usage_pct_col = "CPU usage [%]"

memory_capacity_col = "Memory capacity provisioned [KB]"
memory_usage_col = "Memory usage [KB]"

disk_read_col = "Disk read throughput [KB/s]"
disk_write_col = "Disk write throughput [KB/s]"

network_rx_col = "Network received throughput [KB/s]"
network_tx_col = "Network transmitted throughput [KB/s]"



# 5. VERIFY REQUIRED COLUMNS


required_columns = [

    timestamp_col,

    cpu_cores_col,
    cpu_capacity_col,
    cpu_usage_mhz_col,
    cpu_usage_pct_col,

    memory_capacity_col,
    memory_usage_col,

    disk_read_col,
    disk_write_col,

    network_rx_col,
    network_tx_col
]


missing_columns = [

    col
    for col in required_columns
    if col not in df.columns

]


if missing_columns:

    raise ValueError(
        "Missing columns:\n"
        + "\n".join(missing_columns)
    )



# 6. CONVERT MEASUREMENTS TO NUMERIC


for col in required_columns:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )


#
# 7. CREATE C_i_req
#
# CPU / COMPUTING REQUIREMENT

# Primary method:
#
# CPU cores × CPU usage %
# Example:
#
# 4 cores × 93.23%
# = 3.73 core-equivalents
#
# but:
# Some Bitbrains VMs report CPU cores = 0.
#
# In those cases we cannot calculate core-equivalent demand
#
# Therefore C_i_req is left NaN for those observations.
# retain CPU usage MHz as the original measured demand.

df["C_i_req_cores"] = np.where(

    df[cpu_cores_col] > 0,

    df[cpu_cores_col]
    *
    (
        df[cpu_usage_pct_col]
        / 100
    ),

    np.nan
)



# 8. CREATE M_i_req
#
# MEMORY REQUIREMENT


# Bitbrains gives memory usage in KB.
# Convert KB -> GB.
# This is directly usable as workload memory demand.

df["M_i_req_GB"] = (

    df[memory_usage_col]

    /

    (1024 ** 2)
)


# 9.  CONVERT PROVISIONED MEMORY TO GB


df["Memory_capacity_GB"] = (

    df[memory_capacity_col]

    /

    (1024 ** 2)
)



# 10. COMPLETENESS CHECK
# For filtering, use measurements that are actually
# present in the trace:
# timestamp
# CPU usage MHz
# CPU usage %
# memory usage
#
#  do NOT require CPU_cores > 0 because that would
# automatically remove VMs whose original trace reports
# zero provisioned cores.

vm_counts = (

    df

    .groupby("VM_ID")

    .agg(

        n_rows=(
            timestamp_col,
            "count"
        ),

        valid_CPU_MHz=(
            cpu_usage_mhz_col,
            "count"
        ),

        valid_CPU_pct=(
            cpu_usage_pct_col,
            "count"
        ),

        valid_memory=(
            memory_usage_col,
            "count"
        )

    )

    .reset_index()
)


# 11. EXPECTED NUMBER OF OBSERVATIONS

# Use the MEDIAN not  the mode.
#
# previous output had:
#
# 8,634 rows for some VMs
# ~16,140 rows for others
#
#  using one global mode as the expected length
# caused completeness values above 100%, e.g. 1.8696.
#
# A completeness score should not be interpreted as
# "187% complete".
#
# Here  check whether required fields are
# populated within each VM's own available trace.

vm_counts["CPU_MHz_completeness"] = (

    vm_counts["valid_CPU_MHz"]

    /

    vm_counts["n_rows"]
)


vm_counts["CPU_pct_completeness"] = (

    vm_counts["valid_CPU_pct"]

    /

    vm_counts["n_rows"]
)


vm_counts["memory_completeness"] = (

    vm_counts["valid_memory"]

    /

    vm_counts["n_rows"]
)


# Overall measurement completeness
vm_counts["completeness"] = (

    vm_counts[
        [
            "CPU_MHz_completeness",
            "CPU_pct_completeness",
            "memory_completeness"
        ]
    ]

    .min(axis=1)
)



# 12. SHOW COMPLETENESS


print("\n" + "=" * 80)
print("COMPLETENESS CHECK")
print("=" * 80)


print(

    vm_counts[
        [
            "VM_ID",
            "n_rows",
            "CPU_MHz_completeness",
            "CPU_pct_completeness",
            "memory_completeness",
            "completeness"
        ]
    ]

    .round(4)

    .to_string(index=False)
)



# 13. FILTER ONLY INCOMPLETE VMs


complete_vms = vm_counts[

    vm_counts["completeness"]
    >= MIN_COMPLETENESS

].copy()


selected_ids = (
    complete_vms["VM_ID"]
    .tolist()
)


print(
    f"\nVMs passing >= "
    f"{MIN_COMPLETENESS:.0%} completeness: "
    f"{len(selected_ids)} / {len(vm_counts)}"
)


print(
    f"Retained VM IDs: "
    f"{selected_ids}"
)



# 14. RETURN TO THE FULL ORIGINAL TIME SERIES


# filter df, NOT the summary table.
# Therefore every timestamp remains.

filtered = df[

    df["VM_ID"].isin(
        selected_ids
    )

].copy()



# 15. KEEP ORIGINAL BITBRAINS STRUCTURE

# Two columns are added at the end:
# C_i_req_cores
# M_i_req_GB

final_columns = [

    "VM_ID",

    "Timestamp [ms]",

    "CPU cores",

    "CPU capacity provisioned [MHZ]",

    "CPU usage [MHZ]",

    "CPU usage [%]",

    "Memory capacity provisioned [KB]",

    "Memory usage [KB]",

    "Disk read throughput [KB/s]",

    "Disk write throughput [KB/s]",

    "Network received throughput [KB/s]",

    "Network transmitted throughput [KB/s]",

    # -------------------------------------------
    # Chapter IV workload parameters
    # -------------------------------------------

    "C_i_req_cores",

    "M_i_req_GB"
]


filtered = filtered[
    final_columns
].copy()



# 16. SORT LIKE THE ORIGINAL FILE


filtered = (

    filtered

    .sort_values(
        by=[
            "VM_ID",
            "Timestamp [ms]"
        ]
    )

    .reset_index(drop=True)
)



# 17. PRINT FIRST 20 ROWS

print("\n" + "=" * 120)
print("FILTERED BITBRAINS DATA — FULL TIME SERIES")
print("=" * 120)


print(

    filtered

    .head(20)

    .to_string(index=False)
)



# 18. CHECK ROWS PER VM


print("\n" + "=" * 80)
print("ROWS RETAINED PER VM")
print("=" * 80)


print(

    filtered

    .groupby("VM_ID")

    .size()

    .to_string()
)


#
# 19. CHECK MISSING VALUES


print("\n" + "=" * 80)
print("MISSING VALUES AFTER FILTERING")
print("=" * 80)


print(

    filtered

    .isna()

    .sum()

    .to_string()
)



# 20. CREATE SEPARATE DIAGNOSTIC SUMMARY

# This summary is NOT  main filtered dataset
# It is only a report showing the workload characteristics

summary = (

    filtered

    .groupby("VM_ID")

    .agg(

        n_rows=(
            "Timestamp [ms]",
            "count"
        ),

        mean_CPU_usage_MHz=(
            "CPU usage [MHZ]",
            "mean"
        ),

        max_CPU_usage_MHz=(
            "CPU usage [MHZ]",
            "max"
        ),

        mean_CPU_usage_pct=(
            "CPU usage [%]",
            "mean"
        ),

        max_CPU_usage_pct=(
            "CPU usage [%]",
            "max"
        ),

        mean_C_i_req_cores=(
            "C_i_req_cores",
            "mean"
        ),

        max_C_i_req_cores=(
            "C_i_req_cores",
            "max"
        ),

        mean_M_i_req_GB=(
            "M_i_req_GB",
            "mean"
        ),

        max_M_i_req_GB=(
            "M_i_req_GB",
            "max"
        )

    )

    .reset_index()
)


print("\n" + "=" * 100)
print("DIAGNOSTIC SUMMARY — NOT THE MAIN DATASET")
print("=" * 100)


print(

    summary

    .round(4)

    .to_string(index=False)
)



# 21. SAVE FULL FILTERED DATA


filtered.to_csv(

    OUTPUT_PATH,

    index=False
)


# Save summary separately
summary.to_csv(

    SUMMARY_PATH,

    index=False
)



# 22. FINAL CONFIRMATION

print("\n" + "=" * 80)
print("FILES SAVED")
print("=" * 80)


print(
    "\nMAIN DATASET:"
)

print(
    OUTPUT_PATH
)

print(
    f"{len(filtered):,} timestamp observations "
    f"from {filtered['VM_ID'].nunique()} VMs."
)


print(
    "\nDIAGNOSTIC SUMMARY:"
)

print(
    SUMMARY_PATH
)
```

```python
# ============================================================
# THESIS DATASET COMBINATION - UPDATED FINAL VERSION
#
# UNIMELB:
# /content/drive/MyDrive/thesis data set
#
# BITBRAINS:
# /content/drive/MyDrive/thesis data set bitbrain
#
# FINAL RESULT:
# 10 UniMelb servers
# +
# 10 Bitbrains workload profiles
# =
# EXACTLY 10 ROWS
#
# IMPORTANT:
# This version DOES NOT create the enormous i × s × t table.
# ============================================================

import pandas as pd
import numpy as np
import os
import glob


# ============================================================
# 0. GOOGLE DRIVE PATHS
# ============================================================

UNIMELB_DIR = "/content/drive/MyDrive/thesis data set"

BITBRAINS_DIR = "/content/drive/MyDrive/thesis data set bitbrain"

OUTPUT_DIR = "/content/drive/MyDrive/THESIS_FINAL_DATASET"

os.makedirs(OUTPUT_DIR, exist_ok=True)


print("=" * 80)
print("THESIS DATASET PREPARATION")
print("=" * 80)

print("\nUniMelb folder:")
print(UNIMELB_DIR)

print("\nBitbrains folder:")
print(BITBRAINS_DIR)

print("\nOutput folder:")
print(OUTPUT_DIR)


# ============================================================
# 1. CHECK FOLDERS
# ============================================================

if not os.path.exists(UNIMELB_DIR):
    raise FileNotFoundError(
        f"UniMelb folder not found:\n{UNIMELB_DIR}"
    )

if not os.path.exists(BITBRAINS_DIR):
    raise FileNotFoundError(
        f"Bitbrains folder not found:\n{BITBRAINS_DIR}"
    )


# ============================================================
# 2. FIND UNIMELB DATASET
# ============================================================

print("\n" + "=" * 80)
print("SEARCHING FOR UNIMELB DATASET")
print("=" * 80)


unimelb_files = glob.glob(
    os.path.join(
        UNIMELB_DIR,
        "*.csv"
    )
)


print(
    f"\nCSV files found in UniMelb folder: "
    f"{len(unimelb_files)}"
)


UNIMELB_SIGNATURE = {
    "Series",
    "CPU_Load",
    "Power",
    "Inlet_Temperature"
}


unimelb_candidates = []


for file in unimelb_files:

    try:

        test = pd.read_csv(
            file,
            nrows=5
        )

        test.columns = (
            test.columns
            .astype(str)
            .str.strip()
        )

        if UNIMELB_SIGNATURE.issubset(
            set(test.columns)
        ):

            unimelb_candidates.append(
                file
            )

            print(
                "✓ UniMelb-compatible:",
                os.path.basename(file)
            )

    except Exception as e:

        print(
            "Skipped:",
            os.path.basename(file),
            "|",
            str(e)[:100]
        )


if len(unimelb_candidates) == 0:

    raise FileNotFoundError(
        "\nNo UniMelb-compatible CSV was found "
        f"in:\n{UNIMELB_DIR}"
    )


# ------------------------------------------------------------
# Prefer your already-filtered 10-server dataset.
#
# NOTE:
# "unfiltered" here means the rows/time-series were not
# summarized yet. It can still already contain the selected
# 10 servers.
# ------------------------------------------------------------

preferred_unimelb_names = [

    "dataset1_10servers_unfiltered.csv",

    "dataset1_10servers.csv",

    "dataset1_combined.csv"
]


UNIMELB_FILE = None


for preferred_name in preferred_unimelb_names:

    for file in unimelb_candidates:

        if (
            os.path.basename(file).lower()
            ==
            preferred_name.lower()
        ):

            UNIMELB_FILE = file
            break

    if UNIMELB_FILE is not None:
        break


if UNIMELB_FILE is None:

    # If none of the known names exist,
    # use the first compatible file.

    UNIMELB_FILE = unimelb_candidates[0]


print("\nSelected UniMelb source:")
print(UNIMELB_FILE)


# ============================================================
# 3. LOAD UNIMELB
# ============================================================

unimelb_raw = pd.read_csv(
    UNIMELB_FILE
)


unimelb_raw.columns = (
    unimelb_raw.columns
    .astype(str)
    .str.strip()
)


print(
    "\nUniMelb original shape:",
    unimelb_raw.shape
)


print(
    "UniMelb unique servers:",
    unimelb_raw["Series"].nunique()
)


# ============================================================
# 4. SELECT EXACTLY 10 UNIMELB SERVERS
# ============================================================

selected_servers = (

    unimelb_raw["Series"]

    .dropna()

    .drop_duplicates()

    .head(10)

    .tolist()
)


if len(selected_servers) < 10:

    raise ValueError(
        f"\nOnly {len(selected_servers)} UniMelb "
        "servers were found. Need 10."
    )


unimelb = (

    unimelb_raw[

        unimelb_raw["Series"].isin(
            selected_servers
        )

    ]

    .copy()
)


# Simple server IDs

server_map = {

    server: i + 1

    for i, server
    in enumerate(selected_servers)

}


unimelb["server_id"] = (

    unimelb["Series"]

    .map(server_map)

)


print("\nSelected UniMelb servers:")


for server, sid in server_map.items():

    print(
        f"Server {sid}: {server}"
    )


# ============================================================
# 5. FIND BITBRAINS FILES 1.csv TO 10.csv ONLY
# ============================================================

print("\n" + "=" * 80)
print("LOADING BITBRAINS FILES 1 TO 10")
print("=" * 80)


bitbrains_parts = []


for vm_id in range(1, 11):

    file = os.path.join(
        BITBRAINS_DIR,
        f"{vm_id}.csv"
    )


    if not os.path.exists(file):

        raise FileNotFoundError(
            f"\nBitbrains file missing:\n{file}\n\n"
            "Expected files 1.csv through 10.csv "
            "inside the Bitbrains Google Drive folder."
        )


    print(
        f"Loading VM {vm_id}: "
        f"{os.path.basename(file)}"
    )


    # Bitbrains original files normally use ;
    # Try semicolon first.

    df = pd.read_csv(
        file,
        sep=";"
    )


    # If it became one giant column,
    # retry normal comma separation.

    if len(df.columns) == 1:

        df = pd.read_csv(
            file
        )


    df.columns = (

        df.columns

        .astype(str)

        .str.strip()
    )


    # Add our own VM ID.
    # This is safer than relying on the
    # original individual files containing VM_ID.

    df["VM_ID"] = vm_id


    bitbrains_parts.append(
        df
    )


# Combine ONLY files 1-10.

bitbrains_raw = pd.concat(
    bitbrains_parts,
    ignore_index=True
)


print(
    "\nBitbrains combined shape:",
    bitbrains_raw.shape
)


print(
    "Bitbrains unique VMs:",
    bitbrains_raw["VM_ID"].nunique()
)


# ============================================================
# 6. CHECK REQUIRED BITBRAINS COLUMNS
# ============================================================

required_bitbrains = [

    "Timestamp [ms]",

    "CPU cores",

    "CPU capacity provisioned [MHZ]",

    "CPU usage [MHZ]",

    "CPU usage [%]",

    "Memory capacity provisioned [KB]",

    "Memory usage [KB]"
]


missing = [

    c for c in required_bitbrains

    if c not in bitbrains_raw.columns
]


if missing:

    print(
        "\nAvailable Bitbrains columns:"
    )

    print(
        bitbrains_raw.columns.tolist()
    )

    raise ValueError(
        "\nMissing required Bitbrains columns:\n"
        + "\n".join(missing)
    )


# ============================================================
# 7. CONVERT IMPORTANT BITBRAINS COLUMNS TO NUMERIC
# ============================================================

numeric_bitbrains = [

    "Timestamp [ms]",

    "CPU cores",

    "CPU capacity provisioned [MHZ]",

    "CPU usage [MHZ]",

    "CPU usage [%]",

    "Memory capacity provisioned [KB]",

    "Memory usage [KB]",

    "Disk read throughput [KB/s]",

    "Disk write throughput [KB/s]",

    "Network received throughput [KB/s]",

    "Network transmitted throughput [KB/s]"
]


for col in numeric_bitbrains:

    if col in bitbrains_raw.columns:

        bitbrains_raw[col] = pd.to_numeric(
            bitbrains_raw[col],
            errors="coerce"
        )


# ============================================================
# 8. DERIVE BITBRAINS WORKLOAD REQUIREMENTS
# ============================================================
#
# C_i_req:
#
# CPU cores × CPU usage fraction
#
# Example:
# 4 cores × 75% = 3 core-equivalents
#
# ============================================================

bitbrains_raw["C_i_req"] = (

    bitbrains_raw["CPU cores"]

    *

    (
        bitbrains_raw["CPU usage [%]"]
        / 100.0
    )
)


# Memory KB -> GB

bitbrains_raw["M_i_req_GB"] = (

    bitbrains_raw["Memory usage [KB]"]

    / 1024.0
    / 1024.0
)


bitbrains_raw[
    "workload_memory_capacity_GB"
] = (

    bitbrains_raw[
        "Memory capacity provisioned [KB]"
    ]

    / 1024.0
    / 1024.0
)


# CPU utilization 0-1

bitbrains_raw[
    "workload_CPU_utilization"
] = (

    bitbrains_raw["CPU usage [%]"]

    / 100.0
)


# Memory utilization 0-1

bitbrains_raw[
    "workload_memory_utilization"
] = np.where(

    bitbrains_raw[
        "Memory capacity provisioned [KB]"
    ] > 0,

    bitbrains_raw[
        "Memory usage [KB]"
    ]

    /

    bitbrains_raw[
        "Memory capacity provisioned [KB]"
    ],

    np.nan
)


# Total workload network traffic

if (
    "Network received throughput [KB/s]"
    in bitbrains_raw.columns
    and
    "Network transmitted throughput [KB/s]"
    in bitbrains_raw.columns
):

    bitbrains_raw[
        "workload_network_total_KBps"
    ] = (

        bitbrains_raw[
            "Network received throughput [KB/s]"
        ]

        +

        bitbrains_raw[
            "Network transmitted throughput [KB/s]"
        ]
    )


# ============================================================
# 9. CONVERT IMPORTANT UNIMELB COLUMNS TO NUMERIC
# ============================================================

unimelb_numeric = [

    "CPU_Load",

    "Power",

    "Network_RX",

    "Network_TX",

    "Inlet_Temperature",

    "CPU1_Temperature",

    "CPU2_Temperature",

    "Fan_speed1",

    "Fan_speed2",

    "Fan_speed3",

    "Fan_speed4",

    "Ram",

    "Ram_Used",

    "No_Of_Running_vms",

    "CPU_cores",

    "CPU_cores_used"
]


for col in unimelb_numeric:

    if col in unimelb.columns:

        unimelb[col] = pd.to_numeric(
            unimelb[col],
            errors="coerce"
        )


# ============================================================
# 10. CREATE UNIMELB DERIVED VALUES
# ============================================================

temp_cols = [

    c for c in [

        "CPU1_Temperature",

        "CPU2_Temperature"

    ]

    if c in unimelb.columns
]


if temp_cols:

    unimelb[
        "CPU_temperature_avg"
    ] = (

        unimelb[temp_cols]

        .mean(axis=1)
    )


fan_cols = [

    c for c in [

        "Fan_speed1",

        "Fan_speed2",

        "Fan_speed3",

        "Fan_speed4"

    ]

    if c in unimelb.columns
]


if fan_cols:

    unimelb[
        "fan_speed_avg"
    ] = (

        unimelb[fan_cols]

        .mean(axis=1)
    )


# Server memory utilization

if (
    "Ram" in unimelb.columns
    and
    "Ram_Used" in unimelb.columns
):

    unimelb[
        "server_memory_utilization"
    ] = np.where(

        unimelb["Ram"] > 0,

        unimelb["Ram_Used"]
        /
        unimelb["Ram"],

        np.nan
    )


# Server total network

if (
    "Network_RX" in unimelb.columns
    and
    "Network_TX" in unimelb.columns
):

    unimelb[
        "server_network_total"
    ] = (

        unimelb["Network_RX"]

        +

        unimelb["Network_TX"]
    )


# ============================================================
# 11. HELPER FUNCTIONS FOR SUMMARIES
# ============================================================

def clean_numeric(series):

    return pd.to_numeric(
        series,
        errors="coerce"
    )


def mean_value(series):

    x = clean_numeric(series)

    return (
        x.mean()
        if x.notna().any()
        else np.nan
    )


def median_value(series):

    x = clean_numeric(series)

    return (
        x.median()
        if x.notna().any()
        else np.nan
    )


def max_value(series):

    x = clean_numeric(series)

    return (
        x.max()
        if x.notna().any()
        else np.nan
    )


def p95_value(series):

    x = clean_numeric(
        series
    ).dropna()

    return (
        x.quantile(0.95)
        if len(x) > 0
        else np.nan
    )


# ============================================================
# 12. SUMMARIZE UNIMELB -> EXACTLY 10 ROWS
# ============================================================

print("\n" + "=" * 80)
print("CREATING 10 SERVER PROFILES")
print("=" * 80)


server_rows = []


for server_id, g in unimelb.groupby(
    "server_id"
):

    row = {}


    row["server_id"] = int(
        server_id
    )


    row["server_name"] = (
        g["Series"]
        .dropna()
        .astype(str)
        .iloc[0]
    )


    # ----------------------------------------
    # CPU CAPACITY
    # ----------------------------------------

    if "CPU_cores" in g.columns:

        row["C_s"] = max_value(
            g["CPU_cores"]
        )


    # ----------------------------------------
    # CPU LOAD
    # ----------------------------------------

    if "CPU_Load" in g.columns:

        row[
            "CPU_load_mean"
        ] = mean_value(
            g["CPU_Load"]
        )

        row[
            "CPU_load_p95"
        ] = p95_value(
            g["CPU_Load"]
        )

        row[
            "CPU_load_max"
        ] = max_value(
            g["CPU_Load"]
        )


        # Relative normalized CPU load
        x = clean_numeric(
            g["CPU_Load"]
        )

        xmin = x.min()
        xmax = x.max()


        if (
            pd.notna(xmin)
            and
            pd.notna(xmax)
            and
            xmax > xmin
        ):

            normalized = (
                (x - xmin)
                /
                (xmax - xmin)
            )

            row[
                "CPU_load_normalized_mean"
            ] = normalized.mean()

        else:

            row[
                "CPU_load_normalized_mean"
            ] = 0.0


    # ----------------------------------------
    # RAM
    # ----------------------------------------

    if "Ram" in g.columns:

        row["M_s_raw"] = max_value(
            g["Ram"]
        )


    if "Ram_Used" in g.columns:

        row[
            "RAM_used_mean"
        ] = mean_value(
            g["Ram_Used"]
        )

        row[
            "RAM_used_p95"
        ] = p95_value(
            g["Ram_Used"]
        )


    if (
        "server_memory_utilization"
        in g.columns
    ):

        valid_memory = (

            g[
                "server_memory_utilization"
            ]

            .where(
                (
                    g[
                        "server_memory_utilization"
                    ] >= 0
                )

                &

                (
                    g[
                        "server_memory_utilization"
                    ] <= 1
                )
            )
        )


        row[
            "server_memory_utilization_mean"
        ] = valid_memory.mean()


        row[
            "server_memory_utilization_p95"
        ] = valid_memory.quantile(
            0.95
        )


    # ----------------------------------------
    # POWER
    # ----------------------------------------

    if "Power" in g.columns:

        row[
            "P_s_mean"
        ] = mean_value(
            g["Power"]
        )

        row[
            "P_s_median"
        ] = median_value(
            g["Power"]
        )

        row[
            "P_s_p95"
        ] = p95_value(
            g["Power"]
        )

        row[
            "P_s_max"
        ] = max_value(
            g["Power"]
        )


    # ----------------------------------------
    # INLET TEMPERATURE
    # ----------------------------------------

    if "Inlet_Temperature" in g.columns:

        row[
            "inlet_temperature_mean"
        ] = mean_value(
            g["Inlet_Temperature"]
        )

        row[
            "inlet_temperature_p95"
        ] = p95_value(
            g["Inlet_Temperature"]
        )


    # ----------------------------------------
    # CPU TEMPERATURE
    # ----------------------------------------

    if "CPU_temperature_avg" in g.columns:

        row[
            "CPU_temperature_mean"
        ] = mean_value(
            g["CPU_temperature_avg"]
        )

        row[
            "CPU_temperature_p95"
        ] = p95_value(
            g["CPU_temperature_avg"]
        )

        row[
            "CPU_temperature_max"
        ] = max_value(
            g["CPU_temperature_avg"]
        )


    # ----------------------------------------
    # FAN
    # ----------------------------------------

    if "fan_speed_avg" in g.columns:

        row[
            "fan_speed_mean"
        ] = mean_value(
            g["fan_speed_avg"]
        )


    # ----------------------------------------
    # NETWORK
    # ----------------------------------------

    if "server_network_total" in g.columns:

        row[
            "server_network_mean"
        ] = mean_value(
            g["server_network_total"]
        )

        row[
            "server_network_p95"
        ] = p95_value(
            g["server_network_total"]
        )


    # ----------------------------------------
    # RUNNING VMs
    # ----------------------------------------

    if "No_Of_Running_vms" in g.columns:

        row[
            "running_VMs_mean"
        ] = mean_value(
            g["No_Of_Running_vms"]
        )

        row[
            "running_VMs_max"
        ] = max_value(
            g["No_Of_Running_vms"]
        )


    row[
        "server_observations"
    ] = len(g)


    server_rows.append(
        row
    )


server_summary = pd.DataFrame(
    server_rows
)


server_summary = (

    server_summary

    .sort_values(
        "server_id"
    )

    .reset_index(
        drop=True
    )
)


if len(server_summary) != 10:

    raise ValueError(
        f"\nExpected 10 server rows, "
        f"got {len(server_summary)}."
    )


print(
    "\n✓ UniMelb summary:",
    len(server_summary),
    "rows"
)


# ============================================================
# 13. SUMMARIZE BITBRAINS -> EXACTLY 10 ROWS
# ============================================================

print("\n" + "=" * 80)
print("CREATING 10 WORKLOAD PROFILES")
print("=" * 80)


workload_rows = []


for vm_id, g in bitbrains_raw.groupby(
    "VM_ID"
):

    row = {}


    row[
        "workload_id"
    ] = int(vm_id)


    row[
        "original_VM_ID"
    ] = int(vm_id)


    # ----------------------------------------
    # CPU REQUIREMENT
    # ----------------------------------------

    row[
        "C_i_req_mean"
    ] = mean_value(
        g["C_i_req"]
    )


    row[
        "C_i_req_p95"
    ] = p95_value(
        g["C_i_req"]
    )


    row[
        "C_i_req_max"
    ] = max_value(
        g["C_i_req"]
    )


    # ----------------------------------------
    # CPU UTILIZATION
    # ----------------------------------------

    row[
        "workload_CPU_utilization_mean"
    ] = mean_value(
        g[
            "workload_CPU_utilization"
        ]
    )


    row[
        "workload_CPU_utilization_p95"
    ] = p95_value(
        g[
            "workload_CPU_utilization"
        ]
    )


    # ----------------------------------------
    # CPU MHz
    # ----------------------------------------

    row[
        "CPU_usage_MHz_mean"
    ] = mean_value(
        g["CPU usage [MHZ]"]
    )


    row[
        "CPU_usage_MHz_p95"
    ] = p95_value(
        g["CPU usage [MHZ]"]
    )


    # ----------------------------------------
    # MEMORY REQUIREMENT
    # ----------------------------------------

    row[
        "M_i_req_GB_mean"
    ] = mean_value(
        g["M_i_req_GB"]
    )


    row[
        "M_i_req_GB_p95"
    ] = p95_value(
        g["M_i_req_GB"]
    )


    row[
        "M_i_req_GB_max"
    ] = max_value(
        g["M_i_req_GB"]
    )


    # ----------------------------------------
    # MEMORY UTILIZATION
    # ----------------------------------------

    row[
        "workload_memory_utilization_mean"
    ] = mean_value(
        g[
            "workload_memory_utilization"
        ]
    )


    row[
        "workload_memory_utilization_p95"
    ] = p95_value(
        g[
            "workload_memory_utilization"
        ]
    )


    # ----------------------------------------
    # NETWORK
    # ----------------------------------------

    if (
        "workload_network_total_KBps"
        in g.columns
    ):

        row[
            "workload_network_KBps_mean"
        ] = mean_value(
            g[
                "workload_network_total_KBps"
            ]
        )

        row[
            "workload_network_KBps_p95"
        ] = p95_value(
            g[
                "workload_network_total_KBps"
            ]
        )


    # ----------------------------------------
    # DISK
    # ----------------------------------------

    if (
        "Disk read throughput [KB/s]"
        in g.columns
    ):

        row[
            "disk_read_KBps_mean"
        ] = mean_value(
            g[
                "Disk read throughput [KB/s]"
            ]
        )


    if (
        "Disk write throughput [KB/s]"
        in g.columns
    ):

        row[
            "disk_write_KBps_mean"
        ] = mean_value(
            g[
                "Disk write throughput [KB/s]"
            ]
        )


    row[
        "workload_observations"
    ] = len(g)


    workload_rows.append(
        row
    )


workload_summary = pd.DataFrame(
    workload_rows
)


workload_summary = (

    workload_summary

    .sort_values(
        "workload_id"
    )

    .reset_index(
        drop=True
    )
)


if len(workload_summary) != 10:

    raise ValueError(
        f"\nExpected 10 Bitbrains workload rows, "
        f"got {len(workload_summary)}."
    )


print(
    "\n✓ Bitbrains summary:",
    len(workload_summary),
    "rows"
)


# ============================================================
# 14. PAIR 10 WORKLOADS WITH 10 SERVERS
# ============================================================
#
# Row 1:
# Server 1 + Workload 1
#
# Row 2:
# Server 2 + Workload 2
#
# ...
#
# Row 10:
# Server 10 + Workload 10
#
# IMPORTANT:
# This is an EXPERIMENTAL pairing.
# It does not claim VM 1 historically ran on Server 1.
# ============================================================


server_summary[
    "workload_id"
] = np.arange(
    1,
    11
)


final_dataset = pd.merge(

    server_summary,

    workload_summary,

    on="workload_id",

    how="left",

    validate="one_to_one"
)


# ============================================================
# 15. ADD SOURCE / DOCUMENTATION COLUMNS
# ============================================================

final_dataset[
    "server_data_source"
] = "UniMelb"


final_dataset[
    "workload_data_source"
] = "Bitbrains"


final_dataset[
    "pairing_type"
] = "Experimental simulation pairing"


# ============================================================
# 16. REORDER IMPORTANT COLUMNS
# ============================================================

front = [

    "server_id",

    "server_name",

    "workload_id",

    "original_VM_ID",

    "server_data_source",

    "workload_data_source",

    "pairing_type",

    "C_s",

    "C_i_req_mean",

    "C_i_req_p95",

    "C_i_req_max",

    "M_s_raw",

    "M_i_req_GB_mean",

    "M_i_req_GB_p95",

    "M_i_req_GB_max",

    "CPU_load_mean",

    "CPU_load_p95",

    "CPU_load_max",

    "CPU_load_normalized_mean",

    "workload_CPU_utilization_mean",

    "workload_CPU_utilization_p95",

    "server_memory_utilization_mean",

    "server_memory_utilization_p95",

    "workload_memory_utilization_mean",

    "workload_memory_utilization_p95",

    "P_s_mean",

    "P_s_median",

    "P_s_p95",

    "P_s_max",

    "inlet_temperature_mean",

    "inlet_temperature_p95",

    "CPU_temperature_mean",

    "CPU_temperature_p95",

    "CPU_temperature_max",

    "fan_speed_mean",

    "server_network_mean",

    "server_network_p95",

    "workload_network_KBps_mean",

    "workload_network_KBps_p95",

    "disk_read_KBps_mean",

    "disk_write_KBps_mean",

    "running_VMs_mean",

    "running_VMs_max",

    "server_observations",

    "workload_observations"
]


front = [

    c for c in front

    if c in final_dataset.columns
]


remaining = [

    c for c in final_dataset.columns

    if c not in front
]


final_dataset = final_dataset[
    front + remaining
]


# ============================================================
# 17. FINAL QUALITY CHECK
# ============================================================

print("\n" + "=" * 80)
print("FINAL QUALITY CHECK")
print("=" * 80)


print(
    "\nFinal rows:",
    len(final_dataset)
)


print(
    "Unique servers:",
    final_dataset[
        "server_id"
    ].nunique()
)


print(
    "Unique workloads:",
    final_dataset[
        "workload_id"
    ].nunique()
)


print(
    "Unique original Bitbrains VMs:",
    final_dataset[
        "original_VM_ID"
    ].nunique()
)


if (
    len(final_dataset) == 10
    and
    final_dataset["server_id"].nunique() == 10
    and
    final_dataset["workload_id"].nunique() == 10
):

    print(
        "\n✓ SUCCESS"
    )

    print(
        "10 servers + 10 workload profiles "
        "= 10 final rows."
    )

else:

    raise ValueError(
        "Final dataset failed the 10 × 10 check."
    )


# ============================================================
# 18. SAVE ONLY SMALL SUMMARY FILES
# ============================================================
#
# This fixes the previous KeyboardInterrupt problem.
#
# We are NOT writing a ~1 million row cross-product.
# ============================================================

SERVER_OUTPUT = os.path.join(
    OUTPUT_DIR,
    "01_UNIMELB_10_SERVER_SUMMARY.csv"
)


WORKLOAD_OUTPUT = os.path.join(
    OUTPUT_DIR,
    "02_BITBRAINS_10_WORKLOAD_SUMMARY.csv"
)


FINAL_OUTPUT = os.path.join(
    OUTPUT_DIR,
    "03_FINAL_10_SERVER_10_WORKLOAD_DATASET.csv"
)


server_summary.to_csv(
    SERVER_OUTPUT,
    index=False
)


workload_summary.to_csv(
    WORKLOAD_OUTPUT,
    index=False
)


final_dataset.to_csv(
    FINAL_OUTPUT,
    index=False
)


# ============================================================
# 19. DISPLAY FINAL DATASET
# ============================================================

print("\n" + "=" * 80)
print("FINAL 10-ROW DATASET")
print("=" * 80)


display(
    final_dataset
)


# ============================================================
# 20. OUTPUT PATHS
# ============================================================

print("\n" + "=" * 80)
print("FILES SAVED SUCCESSFULLY")
print("=" * 80)


print(
    "\n1. UniMelb 10-server summary:"
)

print(
    SERVER_OUTPUT
)


print(
    "\n2. Bitbrains 10-workload summary:"
)

print(
    WORKLOAD_OUTPUT
)


print(
    "\n3. FINAL combined 10-row dataset:"
)

print(
    FINAL_OUTPUT
)
```

```python
# ============================================================
# FIX: REBUILD BITBRAINS WORKLOAD SUMMARY USING FILES 1-10
#
# This replaces the OLD 7-workload workload_summary.
#
# Bitbrains folder:
# /content/drive/MyDrive/thesis data set bitbrain
#
# OUTPUT:
# workload_summary = EXACTLY 10 rows
# ============================================================

import pandas as pd
import numpy as np
import os


BITBRAINS_DIR = (
    "/content/drive/MyDrive/thesis data set bitbrain"
)


print("=" * 80)
print("REBUILDING BITBRAINS USING FILES 1.csv TO 10.csv")
print("=" * 80)


# ============================================================
# 1. LOAD EXACTLY FILES 1.csv TO 10.csv
# ============================================================

bitbrains_parts = []


for vm_id in range(1, 11):

    file_path = os.path.join(
        BITBRAINS_DIR,
        f"{vm_id}.csv"
    )


    print(
        f"\nChecking VM {vm_id}:",
        file_path
    )


    if not os.path.exists(file_path):

        raise FileNotFoundError(
            f"\n❌ FILE NOT FOUND:\n"
            f"{file_path}\n\n"
            f"Make sure {vm_id}.csv is inside:\n"
            f"{BITBRAINS_DIR}"
        )


    # --------------------------------------------------------
    # Bitbrains normally uses semicolon separator
    # --------------------------------------------------------

    df = pd.read_csv(
        file_path,
        sep=";"
    )


    # If pandas reads everything as one column,
    # retry with comma separator.

    if len(df.columns) == 1:

        df = pd.read_csv(
            file_path
        )


    # Clean column names

    df.columns = (
        df.columns
        .astype(str)
        .str.strip()
    )


    # Add our own VM identifier

    df["VM_ID"] = vm_id


    print(
        f"✓ Loaded VM {vm_id}"
        f" | rows = {len(df)}"
        f" | columns = {len(df.columns)}"
    )


    bitbrains_parts.append(
        df
    )


# ============================================================
# 2. COMBINE THE 10 BITBRAINS FILES
# ============================================================

bitbrains_raw = pd.concat(
    bitbrains_parts,
    ignore_index=True
)


print("\n" + "=" * 80)
print("BITBRAINS COMBINATION CHECK")
print("=" * 80)


print(
    "\nTotal rows:",
    len(bitbrains_raw)
)


print(
    "Unique VM_IDs:",
    bitbrains_raw["VM_ID"].nunique()
)


print(
    "VM IDs found:",
    sorted(
        bitbrains_raw[
            "VM_ID"
        ].unique()
    )
)


# ============================================================
# 3. REQUIRED COLUMNS
# ============================================================

required_columns = [

    "Timestamp [ms]",

    "CPU cores",

    "CPU capacity provisioned [MHZ]",

    "CPU usage [MHZ]",

    "CPU usage [%]",

    "Memory capacity provisioned [KB]",

    "Memory usage [KB]"
]


missing_columns = [

    col
    for col in required_columns

    if col not in bitbrains_raw.columns
]


if missing_columns:

    print(
        "\nAvailable columns:"
    )

    print(
        bitbrains_raw.columns.tolist()
    )


    raise ValueError(
        "\n❌ Missing required columns:\n"
        +
        "\n".join(missing_columns)
    )


# ============================================================
# 4. CONVERT NUMERIC COLUMNS
# ============================================================

numeric_columns = [

    "Timestamp [ms]",

    "CPU cores",

    "CPU capacity provisioned [MHZ]",

    "CPU usage [MHZ]",

    "CPU usage [%]",

    "Memory capacity provisioned [KB]",

    "Memory usage [KB]",

    "Disk read throughput [KB/s]",

    "Disk write throughput [KB/s]",

    "Network received throughput [KB/s]",

    "Network transmitted throughput [KB/s]"
]


for col in numeric_columns:

    if col in bitbrains_raw.columns:

        bitbrains_raw[col] = pd.to_numeric(
            bitbrains_raw[col],
            errors="coerce"
        )


# ============================================================
# 5. DERIVE STANDARDIZED CPU VARIABLES
# ============================================================

# ------------------------------------------------------------
# CPU requirement in core-equivalent units
#
# Example:
#
# 4 cores × 75% utilization
# = 3 core-equivalent CPU requirement
# ------------------------------------------------------------

bitbrains_raw[
    "C_i_req"
] = (

    bitbrains_raw[
        "CPU cores"
    ]

    *

    (
        bitbrains_raw[
            "CPU usage [%]"
        ]

        / 100.0
    )
)


# CPU utilization:
#
# 93% -> 0.93

bitbrains_raw[
    "workload_CPU_utilization"
] = (

    bitbrains_raw[
        "CPU usage [%]"
    ]

    / 100.0
)


bitbrains_raw[
    "workload_CPU_utilization"
] = (

    bitbrains_raw[
        "workload_CPU_utilization"
    ]

    .clip(
        lower=0,
        upper=1
    )
)


# ============================================================
# 6. DERIVE STANDARDIZED MEMORY VARIABLES
# ============================================================

# ------------------------------------------------------------
# KB -> GB
# ------------------------------------------------------------

bitbrains_raw[
    "M_i_req_GB"
] = (

    bitbrains_raw[
        "Memory usage [KB]"
    ]

    / 1024.0
    / 1024.0
)


bitbrains_raw[
    "workload_memory_capacity_GB"
] = (

    bitbrains_raw[
        "Memory capacity provisioned [KB]"
    ]

    / 1024.0
    / 1024.0
)


# ------------------------------------------------------------
# Memory utilization
# ------------------------------------------------------------

bitbrains_raw[
    "workload_memory_utilization"
] = np.where(

    bitbrains_raw[
        "Memory capacity provisioned [KB]"
    ] > 0,

    bitbrains_raw[
        "Memory usage [KB]"
    ]

    /

    bitbrains_raw[
        "Memory capacity provisioned [KB]"
    ],

    np.nan
)


bitbrains_raw[
    "workload_memory_utilization"
] = (

    bitbrains_raw[
        "workload_memory_utilization"
    ]

    .clip(
        lower=0,
        upper=1
    )
)


# ============================================================
# 7. NETWORK ACTIVITY
# ============================================================

if (
    "Network received throughput [KB/s]"
    in bitbrains_raw.columns

    and

    "Network transmitted throughput [KB/s]"
    in bitbrains_raw.columns
):

    bitbrains_raw[
        "workload_network_total_KBps"
    ] = (

        bitbrains_raw[
            "Network received throughput [KB/s]"
        ]

        +

        bitbrains_raw[
            "Network transmitted throughput [KB/s]"
        ]
    )


# ============================================================
# 8. HELPER FUNCTIONS
# ============================================================

def clean_numeric(series):

    return pd.to_numeric(
        series,
        errors="coerce"
    )


def mean_value(series):

    x = clean_numeric(series)

    if x.notna().any():
        return x.mean()

    return np.nan


def median_value(series):

    x = clean_numeric(series)

    if x.notna().any():
        return x.median()

    return np.nan


def max_value(series):

    x = clean_numeric(series)

    if x.notna().any():
        return x.max()

    return np.nan


def p95_value(series):

    x = clean_numeric(
        series
    ).dropna()

    if len(x) > 0:
        return x.quantile(0.95)

    return np.nan


# ============================================================
# 9. CREATE EXACTLY ONE ROW PER BITBRAINS WORKLOAD
# ============================================================

workload_rows = []


for vm_id, g in bitbrains_raw.groupby(
    "VM_ID"
):

    row = {}


    # --------------------------------------------------------
    # IDENTIFICATION
    # --------------------------------------------------------

    row[
        "workload_id"
    ] = int(vm_id)


    row[
        "original_VM_ID"
    ] = int(vm_id)


    # --------------------------------------------------------
    # CPU REQUIREMENT
    # --------------------------------------------------------

    row[
        "C_i_req_mean"
    ] = mean_value(
        g["C_i_req"]
    )


    row[
        "C_i_req_p95"
    ] = p95_value(
        g["C_i_req"]
    )


    row[
        "C_i_req_max"
    ] = max_value(
        g["C_i_req"]
    )


    # --------------------------------------------------------
    # CPU UTILIZATION
    # --------------------------------------------------------

    row[
        "workload_CPU_utilization_mean"
    ] = mean_value(
        g[
            "workload_CPU_utilization"
        ]
    )


    row[
        "workload_CPU_utilization_p95"
    ] = p95_value(
        g[
            "workload_CPU_utilization"
        ]
    )


    # --------------------------------------------------------
    # CPU MHz
    # --------------------------------------------------------

    row[
        "CPU_usage_MHz_mean"
    ] = mean_value(
        g[
            "CPU usage [MHZ]"
        ]
    )


    row[
        "CPU_usage_MHz_p95"
    ] = p95_value(
        g[
            "CPU usage [MHZ]"
        ]
    )


    # --------------------------------------------------------
    # MEMORY REQUIREMENT
    # --------------------------------------------------------

    row[
        "M_i_req_GB_mean"
    ] = mean_value(
        g[
            "M_i_req_GB"
        ]
    )


    row[
        "M_i_req_GB_p95"
    ] = p95_value(
        g[
            "M_i_req_GB"
        ]
    )


    row[
        "M_i_req_GB_max"
    ] = max_value(
        g[
            "M_i_req_GB"
        ]
    )


    # --------------------------------------------------------
    # MEMORY UTILIZATION
    # --------------------------------------------------------

    row[
        "workload_memory_utilization_mean"
    ] = mean_value(
        g[
            "workload_memory_utilization"
        ]
    )


    row[
        "workload_memory_utilization_p95"
    ] = p95_value(
        g[
            "workload_memory_utilization"
        ]
    )


    # --------------------------------------------------------
    # NETWORK
    # --------------------------------------------------------

    if (
        "workload_network_total_KBps"
        in g.columns
    ):

        row[
            "workload_network_KBps_mean"
        ] = mean_value(
            g[
                "workload_network_total_KBps"
            ]
        )


        row[
            "workload_network_KBps_p95"
        ] = p95_value(
            g[
                "workload_network_total_KBps"
            ]
        )


    # --------------------------------------------------------
    # DISK READ
    # --------------------------------------------------------

    if (
        "Disk read throughput [KB/s]"
        in g.columns
    ):

        row[
            "disk_read_KBps_mean"
        ] = mean_value(
            g[
                "Disk read throughput [KB/s]"
            ]
        )


    # --------------------------------------------------------
    # DISK WRITE
    # --------------------------------------------------------

    if (
        "Disk write throughput [KB/s]"
        in g.columns
    ):

        row[
            "disk_write_KBps_mean"
        ] = mean_value(
            g[
                "Disk write throughput [KB/s]"
            ]
        )


    # --------------------------------------------------------
    # NUMBER OF ORIGINAL OBSERVATIONS
    # --------------------------------------------------------

    row[
        "workload_observations"
    ] = len(g)


    workload_rows.append(
        row
    )


# ============================================================
# 10. CREATE NEW workload_summary
# ============================================================

workload_summary = pd.DataFrame(
    workload_rows
)


workload_summary = (

    workload_summary

    .sort_values(
        "workload_id"
    )

    .reset_index(
        drop=True
    )
)


# ============================================================
# 11. CRITICAL CHECK
# ============================================================

print("\n" + "=" * 80)
print("NEW WORKLOAD SUMMARY CHECK")
print("=" * 80)


print(
    "\nNumber of workload profiles:",
    len(workload_summary)
)


print(
    "Unique workload IDs:",
    workload_summary[
        "workload_id"
    ].nunique()
)


print(
    "Original VM IDs:"
)


print(
    workload_summary[
        "original_VM_ID"
    ].tolist()
)


if (
    len(workload_summary) != 10

    or

    workload_summary[
        "workload_id"
    ].nunique() != 10
):

    raise ValueError(
        "\n❌ Still did not create 10 workload profiles.\n"
        "Check whether files 1.csv through 10.csv "
        "are actually present and contain valid data."
    )


print(
    "\n✓ SUCCESS!"
)

print(
    "The OLD 7-workload workload_summary "
    "has now been replaced."
)

print(
    "New workload_summary contains "
    "10 unique Bitbrains workloads."
)


# ============================================================
# 12. DISPLAY THE 10 WORKLOADS
# ============================================================

display(
    workload_summary
)


# ============================================================
# 13. SAVE NEW 10-WORKLOAD SUMMARY
# ============================================================

OUTPUT_DIR = (
    "/content/drive/MyDrive/"
    "THESIS_FINAL_DATASET"
)


os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)


OUTPUT_FILE = os.path.join(
    OUTPUT_DIR,
    "BITBRAINS_10_WORKLOAD_SUMMARY.csv"
)


workload_summary.to_csv(
    OUTPUT_FILE,
    index=False
)


print("\nSaved to:")
print(OUTPUT_FILE)


print("\n" + "=" * 80)
print("NEXT STEP")
print("=" * 80)


print(
    "\nNow rerun the FINAL STANDARDIZATION "
    "10 SERVERS + 10 WORKLOADS code."
)

print(
    "\nIt should now begin with:"
)

print(
    "Server profiles: 10"
)

print(
    "Workload profiles: 10"
)
```

```python
# ============================================================
# FINAL STANDARDIZATION — 10 UNIMELB SERVERS + 10 BITBRAINS VMs
#
# RUN THIS AFTER THE UPDATED 10-WORKLOAD COMBINATION CODE
#
# Expected existing DataFrames:
#     server_summary
#     workload_summary
#     final_dataset
#
# FINAL OUTPUT:
#     10 servers
#     10 workloads
#     10 paired rows
# ============================================================

import pandas as pd
import numpy as np
import os


print("=" * 80)
print("FINAL STANDARDIZATION — 10 SERVERS + 10 WORKLOADS")
print("=" * 80)


# ============================================================
# 0. CHECK PREVIOUS CODE
# ============================================================

required_dfs = [
    "server_summary",
    "workload_summary",
    "final_dataset"
]

for df_name in required_dfs:

    if df_name not in globals():

        raise NameError(
            f"\n'{df_name}' does not exist.\n\n"
            "Please run the UPDATED 10-server + "
            "10-Bitbrains-workload combination code first."
        )


servers = server_summary.copy()
workloads = workload_summary.copy()


# ============================================================
# 1. BASIC CHECK
# ============================================================

print("\n" + "=" * 80)
print("INPUT CHECK")
print("=" * 80)

print(
    "\nServer profiles:",
    len(servers)
)

print(
    "Workload profiles:",
    len(workloads)
)


if len(servers) != 10:

    raise ValueError(
        f"Expected 10 server profiles, got {len(servers)}."
    )


if len(workloads) != 10:

    raise ValueError(
        f"Expected 10 workload profiles, got {len(workloads)}."
    )


if servers["server_id"].nunique() != 10:

    raise ValueError(
        "server_id does not contain 10 unique servers."
    )


if workloads["workload_id"].nunique() != 10:

    raise ValueError(
        "workload_id does not contain 10 unique workloads."
    )


print(
    "\n✓ Exactly 10 servers and 10 workloads found."
)


# ============================================================
# 2. CPU STANDARDIZATION — WORKLOAD
# ============================================================

print("\n" + "=" * 80)
print("1. CPU STANDARDIZATION")
print("=" * 80)


# ------------------------------------------------------------
# Workload CPU utilization
#
# Already expressed as 0-1 in the previous code.
# We verify and clip values to a valid range.
# ------------------------------------------------------------

for col in [
    "workload_CPU_utilization_mean",
    "workload_CPU_utilization_p95"
]:

    if col in workloads.columns:

        workloads[col] = pd.to_numeric(
            workloads[col],
            errors="coerce"
        )

        workloads[col] = workloads[col].clip(
            lower=0,
            upper=1
        )


# ------------------------------------------------------------
# Workload CPU requirement
#
# C_i_req was originally:
#
# CPU cores × CPU usage %
#
# We retain:
#
# mean = typical demand
# p95  = high demand
# max  = observed extreme
# ------------------------------------------------------------

for col in [
    "C_i_req_mean",
    "C_i_req_p95",
    "C_i_req_max"
]:

    if col in workloads.columns:

        workloads[col] = pd.to_numeric(
            workloads[col],
            errors="coerce"
        )


print(
    "\n✓ Bitbrains CPU utilization retained on 0-1 scale."
)

print(
    "✓ CPU requirement retained as "
    "mean, P95 and maximum."
)


# ============================================================
# 3. CPU STANDARDIZATION — SERVER
# ============================================================

# ------------------------------------------------------------
# C_s = server CPU capacity
# ------------------------------------------------------------

if "C_s" in servers.columns:

    servers["C_s"] = pd.to_numeric(
        servers["C_s"],
        errors="coerce"
    )


# ------------------------------------------------------------
# CPU load normalized mean
#
# This is a RELATIVE 0-1 indicator.
#
# It does NOT replace physical CPU capacity C_s.
# ------------------------------------------------------------

if "CPU_load_normalized_mean" in servers.columns:

    servers[
        "CPU_load_normalized_mean"
    ] = pd.to_numeric(
        servers[
            "CPU_load_normalized_mean"
        ],
        errors="coerce"
    )

    servers[
        "CPU_load_normalized_mean"
    ] = servers[
        "CPU_load_normalized_mean"
    ].clip(
        lower=0,
        upper=1
    )


print(
    "✓ UniMelb CPU capacity C_s retained."
)

print(
    "✓ Relative CPU load kept separately "
    "from physical CPU capacity."
)


# ============================================================
# 4. MEMORY STANDARDIZATION — WORKLOAD
# ============================================================

print("\n" + "=" * 80)
print("2. MEMORY STANDARDIZATION")
print("=" * 80)


# ------------------------------------------------------------
# Bitbrains workload memory is already GB.
# ------------------------------------------------------------

for col in [
    "M_i_req_GB_mean",
    "M_i_req_GB_p95",
    "M_i_req_GB_max"
]:

    if col in workloads.columns:

        workloads[col] = pd.to_numeric(
            workloads[col],
            errors="coerce"
        )


# ------------------------------------------------------------
# Workload memory utilization 0-1
# ------------------------------------------------------------

for col in [
    "workload_memory_utilization_mean",
    "workload_memory_utilization_p95"
]:

    if col in workloads.columns:

        workloads[col] = pd.to_numeric(
            workloads[col],
            errors="coerce"
        )

        workloads[col] = workloads[col].clip(
            lower=0,
            upper=1
        )


print(
    "\n✓ Bitbrains memory requirement is in GB."
)

print(
    "✓ Workload memory utilization is on 0-1 scale."
)


# ============================================================
# 5. MEMORY STANDARDIZATION — SERVER
# ============================================================

# ------------------------------------------------------------
# IMPORTANT:
#
# M_s_raw remains RAW until the physical unit of
# UniMelb Ram is verified.
#
# Do NOT rename it M_s_GB yet.
# ------------------------------------------------------------

if "M_s_raw" in servers.columns:

    servers[
        "M_s_raw"
    ] = pd.to_numeric(
        servers[
            "M_s_raw"
        ],
        errors="coerce"
    )


for col in [
    "server_memory_utilization_mean",
    "server_memory_utilization_p95"
]:

    if col in servers.columns:

        servers[col] = pd.to_numeric(
            servers[col],
            errors="coerce"
        )

        # Invalid utilization values are not forced.
        servers[col] = servers[col].where(
            (servers[col] >= 0)
            &
            (servers[col] <= 1)
        )


print(
    "✓ UniMelb RAM retained as M_s_raw."
)

print(
    "⚠ M_s_raw is NOT treated as GB until "
    "the UniMelb RAM unit is verified."
)


# ============================================================
# 6. POWER STANDARDIZATION
# ============================================================

print("\n" + "=" * 80)
print("3. POWER STANDARDIZATION")
print("=" * 80)


power_columns = [
    "P_s_mean",
    "P_s_median",
    "P_s_p95",
    "P_s_max"
]


for col in power_columns:

    if col in servers.columns:

        servers[col] = pd.to_numeric(
            servers[col],
            errors="coerce"
        )


# ------------------------------------------------------------
# Relative power indicator across the 10 servers
#
# NOTE:
# This does NOT replace actual power.
# ------------------------------------------------------------

if "P_s_mean" in servers.columns:

    p_min = servers[
        "P_s_mean"
    ].min()

    p_max = servers[
        "P_s_mean"
    ].max()


    if (
        pd.notna(p_min)
        and
        pd.notna(p_max)
        and
        p_max > p_min
    ):

        servers[
            "P_s_mean_normalized"
        ] = (

            servers[
                "P_s_mean"
            ]
            -
            p_min

        ) / (
            p_max
            -
            p_min
        )

    else:

        servers[
            "P_s_mean_normalized"
        ] = 0.0


print(
    "\n✓ Original power values retained."
)

print(
    "✓ Additional 0-1 relative power "
    "indicator created."
)


# ============================================================
# 7. TEMPERATURE STANDARDIZATION
# ============================================================

print("\n" + "=" * 80)
print("4. TEMPERATURE STANDARDIZATION")
print("=" * 80)


temperature_columns = [

    "inlet_temperature_mean",

    "inlet_temperature_p95",

    "CPU_temperature_mean",

    "CPU_temperature_p95",

    "CPU_temperature_max"
]


for col in temperature_columns:

    if col in servers.columns:

        servers[col] = pd.to_numeric(
            servers[col],
            errors="coerce"
        )


# ------------------------------------------------------------
# Relative temperature indicator across servers
# ------------------------------------------------------------

if "CPU_temperature_mean" in servers.columns:

    t_min = servers[
        "CPU_temperature_mean"
    ].min()

    t_max = servers[
        "CPU_temperature_mean"
    ].max()


    if (
        pd.notna(t_min)
        and
        pd.notna(t_max)
        and
        t_max > t_min
    ):

        servers[
            "CPU_temperature_normalized"
        ] = (

            servers[
                "CPU_temperature_mean"
            ]
            -
            t_min

        ) / (
            t_max
            -
            t_min
        )

    else:

        servers[
            "CPU_temperature_normalized"
        ] = 0.0


print(
    "\n✓ Original temperature retained."
)

print(
    "✓ Additional relative 0-1 temperature "
    "indicator created."
)


# ============================================================
# 8. NETWORK STANDARDIZATION
# ============================================================

print("\n" + "=" * 80)
print("5. NETWORK STANDARDIZATION")
print("=" * 80)


# ------------------------------------------------------------
# Workload network
# ------------------------------------------------------------

if "workload_network_KBps_mean" in workloads.columns:

    workloads[
        "workload_network_KBps_mean"
    ] = pd.to_numeric(
        workloads[
            "workload_network_KBps_mean"
        ],
        errors="coerce"
    )


    w_min = workloads[
        "workload_network_KBps_mean"
    ].min()

    w_max = workloads[
        "workload_network_KBps_mean"
    ].max()


    if (
        pd.notna(w_min)
        and
        pd.notna(w_max)
        and
        w_max > w_min
    ):

        workloads[
            "workload_network_normalized"
        ] = (

            workloads[
                "workload_network_KBps_mean"
            ]
            -
            w_min

        ) / (
            w_max
            -
            w_min
        )

    else:

        workloads[
            "workload_network_normalized"
        ] = 0.0


# ------------------------------------------------------------
# Server network
# ------------------------------------------------------------

if "server_network_mean" in servers.columns:

    servers[
        "server_network_mean"
    ] = pd.to_numeric(
        servers[
            "server_network_mean"
        ],
        errors="coerce"
    )


    s_min = servers[
        "server_network_mean"
    ].min()

    s_max = servers[
        "server_network_mean"
    ].max()


    if (
        pd.notna(s_min)
        and
        pd.notna(s_max)
        and
        s_max > s_min
    ):

        servers[
            "server_network_normalized"
        ] = (

            servers[
                "server_network_mean"
            ]
            -
            s_min

        ) / (
            s_max
            -
            s_min
        )

    else:

        servers[
            "server_network_normalized"
        ] = 0.0


print(
    "\n✓ Network physical values retained."
)

print(
    "✓ Separate relative 0-1 indicators created."
)


# ============================================================
# 9. CREATE MODEL PARAMETER COLUMNS
# ============================================================

print("\n" + "=" * 80)
print("6. MODEL PARAMETER SELECTION")
print("=" * 80)


# ------------------------------------------------------------
# IMPORTANT:
#
# We retain mean, P95 and max in the full dataset.
#
# For a conservative capacity requirement table,
# use P95 as the working workload requirement.
#
# This avoids using only the average while also
# avoiding making one single extreme observation
# automatically determine every workload requirement.
# ------------------------------------------------------------

if "C_i_req_p95" in workloads.columns:

    workloads[
        "C_i_req_model"
    ] = workloads[
        "C_i_req_p95"
    ]


if "M_i_req_GB_p95" in workloads.columns:

    workloads[
        "M_i_req_GB_model"
    ] = workloads[
        "M_i_req_GB_p95"
    ]


print(
    "\nWorking CPU requirement:"
)

print(
    "C_i_req_model = C_i_req_p95"
)


print(
    "\nWorking memory requirement:"
)

print(
    "M_i_req_GB_model = M_i_req_GB_p95"
)


# ============================================================
# 10. CREATE FINAL 10 x 10 COMPATIBILITY TABLE
# ============================================================
#
# IMPORTANT:
#
# This is NOT the final 10-row profile table.
#
# This table contains:
#
# 10 workloads × 10 servers = 100 candidate assignments.
#
# That is useful because your optimization needs to know
# which workload could potentially be assigned to which server.
# ============================================================

print("\n" + "=" * 80)
print("7. WORKLOAD-SERVER COMPATIBILITY")
print("=" * 80)


server_capacity_cols = [
    c for c in [
        "server_id",
        "server_name",
        "C_s",
        "M_s_raw"
    ]
    if c in servers.columns
]


workload_requirement_cols = [
    c for c in [
        "workload_id",
        "original_VM_ID",
        "C_i_req_model",
        "M_i_req_GB_model"
    ]
    if c in workloads.columns
]


server_cap = servers[
    server_capacity_cols
].copy()


workload_req = workloads[
    workload_requirement_cols
].copy()


# Cross join:
# every workload with every server

compatibility = workload_req.merge(
    server_cap,
    how="cross"
)


# ------------------------------------------------------------
# CPU feasibility
# ------------------------------------------------------------

if (
    "C_i_req_model"
    in compatibility.columns
    and
    "C_s"
    in compatibility.columns
):

    compatibility[
        "CPU_feasible"
    ] = (

        compatibility[
            "C_i_req_model"
        ]

        <=

        compatibility[
            "C_s"
        ]
    )


# ------------------------------------------------------------
# MEMORY FEASIBILITY
#
# DO NOT calculate yet because M_s_raw is not
# confirmed to be GB.
# ------------------------------------------------------------

compatibility[
    "Memory_feasible"
] = np.nan


compatibility[
    "Memory_check_status"
] = (
    "Pending verification of UniMelb RAM unit"
)


print(
    "\nCandidate workload-server pairs:",
    len(compatibility)
)

print(
    "Expected:",
    10 * 10
)


# ============================================================
# 11. CREATE FINAL 10-ROW STANDARDIZED TABLE
# ============================================================
#
# Workload 1 -> Server 1
# Workload 2 -> Server 2
# ...
# Workload 10 -> Server 10
#
# This pairing is for the SUMMARY PROFILE TABLE only.
#
# It does NOT force the MILP to assign workload 1
# to server 1.
# ============================================================

print("\n" + "=" * 80)
print("8. BUILDING FINAL 10-ROW TABLE")
print("=" * 80)


servers_for_merge = servers.copy()
workloads_for_merge = workloads.copy()


# Ensure both have the same pairing ID

servers_for_merge[
    "pair_id"
] = np.arange(
    1,
    11
)


workloads_for_merge[
    "pair_id"
] = np.arange(
    1,
    11
)


final_standardized = pd.merge(

    servers_for_merge,

    workloads_for_merge,

    on="pair_id",

    how="inner",

    validate="one_to_one",

    suffixes=(
        "_server",
        "_workload"
    )
)


# ============================================================
# 12. ADD DOCUMENTATION
# ============================================================

final_standardized[
    "server_data_source"
] = "UniMelb"


final_standardized[
    "workload_data_source"
] = "Bitbrains"


final_standardized[
    "pairing_type"
] = "Experimental simulation profile pairing"


final_standardized[
    "standardization_note"
] = (
    "Server and workload profiles summarized from "
    "original time-series observations."
)


# ============================================================
# 13. REORDER COLUMNS
# ============================================================

front = [

    "pair_id",

    "server_id",

    "server_name",

    "workload_id",

    "original_VM_ID",

    "server_data_source",

    "workload_data_source",

    "pairing_type",

    "C_s",

    "C_i_req_model",

    "C_i_req_mean",

    "C_i_req_p95",

    "C_i_req_max",

    "M_s_raw",

    "M_i_req_GB_model",

    "M_i_req_GB_mean",

    "M_i_req_GB_p95",

    "M_i_req_GB_max",

    "CPU_load_mean",

    "CPU_load_p95",

    "CPU_load_max",

    "CPU_load_normalized_mean",

    "workload_CPU_utilization_mean",

    "workload_CPU_utilization_p95",

    "server_memory_utilization_mean",

    "server_memory_utilization_p95",

    "workload_memory_utilization_mean",

    "workload_memory_utilization_p95",

    "P_s_mean",

    "P_s_median",

    "P_s_p95",

    "P_s_max",

    "P_s_mean_normalized",

    "inlet_temperature_mean",

    "inlet_temperature_p95",

    "CPU_temperature_mean",

    "CPU_temperature_p95",

    "CPU_temperature_max",

    "CPU_temperature_normalized",

    "fan_speed_mean",

    "server_network_mean",

    "server_network_p95",

    "server_network_normalized",

    "workload_network_KBps_mean",

    "workload_network_KBps_p95",

    "workload_network_normalized",

    "disk_read_KBps_mean",

    "disk_write_KBps_mean",

    "running_VMs_mean",

    "running_VMs_max",

    "server_observations",

    "workload_observations",

    "standardization_note"
]


front = [
    c for c in front
    if c in final_standardized.columns
]


others = [
    c for c in final_standardized.columns
    if c not in front
]


final_standardized = final_standardized[
    front + others
]


# ============================================================
# 14. FINAL QUALITY CHECK
# ============================================================

print("\n" + "=" * 80)
print("FINAL STANDARDIZATION QUALITY CHECK")
print("=" * 80)


print(
    "\nFinal number of rows:",
    len(final_standardized)
)


print(
    "Unique servers:",
    final_standardized[
        "server_id"
    ].nunique()
)


print(
    "Unique workloads:",
    final_standardized[
        "workload_id"
    ].nunique()
)


print(
    "Unique original Bitbrains VMs:",
    final_standardized[
        "original_VM_ID"
    ].nunique()
)


print(
    "\nCompatibility rows:",
    len(compatibility)
)


if (
    len(final_standardized) == 10
    and
    final_standardized[
        "server_id"
    ].nunique() == 10
    and
    final_standardized[
        "workload_id"
    ].nunique() == 10
):

    print(
        "\n✓ SUCCESS:"
    )

    print(
        "Final standardized profile table "
        "contains exactly 10 rows."
    )

    print(
        "Each of the 10 Bitbrains workloads "
        "appears exactly once."
    )

else:

    raise ValueError(
        "\nFinal table is not 10 servers × "
        "10 unique workload profiles."
    )


# ============================================================
# 15. CREATE STANDARDIZATION REPORT
# ============================================================

report = pd.DataFrame({

    "Item": [

        "Number of servers",

        "Number of workloads",

        "CPU workload requirement",

        "CPU workload utilization",

        "Server CPU capacity",

        "Server CPU load",

        "Workload memory",

        "Server memory",

        "Power",

        "Temperature",

        "Network",

        "Final profile table",

        "Compatibility table"
    ],

    "Result": [

        "10 UniMelb servers",

        "10 Bitbrains workloads",

        (
            "C_i_req retained as mean, P95 and max; "
            "P95 selected as working model requirement"
        ),

        (
            "Converted/retained as 0-1 utilization ratio"
        ),

        (
            "C_s retained from UniMelb CPU cores"
        ),

        (
            "Raw load retained plus relative "
            "0-1 normalized indicator"
        ),

        (
            "Bitbrains KB converted to GB; "
            "mean, P95 and max retained"
        ),

        (
            "M_s_raw retained until UniMelb "
            "RAM unit is verified"
        ),

        (
            "Original mean, median, P95 and max retained; "
            "relative 0-1 mean-power indicator added"
        ),

        (
            "Original temperatures retained; "
            "relative 0-1 indicator added"
        ),

        (
            "Physical network measurements retained; "
            "separate relative indicators added"
        ),

        (
            "Exactly 10 rows: "
            "10 servers + 10 unique workload profiles"
        ),

        (
            "100 candidate workload-server pairs "
            "for feasibility checking"
        )
    ]
})


# ============================================================
# 16. SAVE TO GOOGLE DRIVE
# ============================================================

OUTPUT_FOLDER = (
    "/content/drive/MyDrive/"
    "THESIS_FINAL_STANDARDIZED_10_WORKLOADS"
)


os.makedirs(
    OUTPUT_FOLDER,
    exist_ok=True
)


SERVER_FILE = os.path.join(
    OUTPUT_FOLDER,
    "01_UNIMELB_10_SERVERS_STANDARDIZED.csv"
)


WORKLOAD_FILE = os.path.join(
    OUTPUT_FOLDER,
    "02_BITBRAINS_10_WORKLOADS_STANDARDIZED.csv"
)


COMPATIBILITY_FILE = os.path.join(
    OUTPUT_FOLDER,
    "03_WORKLOAD_SERVER_COMPATIBILITY_100_PAIRS.csv"
)


FINAL_FILE = os.path.join(
    OUTPUT_FOLDER,
    "04_FINAL_STANDARDIZED_10_ROWS.csv"
)


REPORT_FILE = os.path.join(
    OUTPUT_FOLDER,
    "05_STANDARDIZATION_REPORT.csv"
)


servers.to_csv(
    SERVER_FILE,
    index=False
)


workloads.to_csv(
    WORKLOAD_FILE,
    index=False
)


compatibility.to_csv(
    COMPATIBILITY_FILE,
    index=False
)


final_standardized.to_csv(
    FINAL_FILE,
    index=False
)


report.to_csv(
    REPORT_FILE,
    index=False
)


# ============================================================
# 17. DISPLAY FINAL TABLE
# ============================================================

print("\n" + "=" * 80)
print("FINAL STANDARDIZED TABLE")
print("=" * 80)


display(
    final_standardized
)


# ============================================================
# 18. SHOW WORKLOAD MAPPING
# ============================================================

print("\n" + "=" * 80)
print("10 SERVER / 10 WORKLOAD PROFILE MAPPING")
print("=" * 80)


mapping_columns = [
    c for c in [
        "server_id",
        "server_name",
        "workload_id",
        "original_VM_ID",
        "C_s",
        "C_i_req_model",
        "M_s_raw",
        "M_i_req_GB_model"
    ]
    if c in final_standardized.columns
]


display(
    final_standardized[
        mapping_columns
    ]
)


# ============================================================
# 19. OUTPUT FILES
# ============================================================

print("\n" + "=" * 80)
print("STANDARDIZATION COMPLETE")
print("=" * 80)


print(
    "\n1. Standardized 10 UniMelb servers:"
)
print(SERVER_FILE)


print(
    "\n2. Standardized 10 Bitbrains workloads:"
)
print(WORKLOAD_FILE)


print(
    "\n3. 100 workload-server candidate pairs:"
)
print(COMPATIBILITY_FILE)


print(
    "\n4. FINAL standardized 10-row table:"
)
print(FINAL_FILE)


print(
    "\n5. Standardization report:"
)
print(REPORT_FILE)


print("\n" + "=" * 80)
print("EXPECTED RESULT")
print("=" * 80)


print(
    "\nServers       = 10"
)

print(
    "Workloads     = 10"
)

print(
    "Final rows    = 10"
)

print(
    "Candidate i-s pairs = 100"
)


print("\n" + "=" * 80)
print("IMPORTANT")
print("=" * 80)


print(
    "\n1. The 10-row table is a SUMMARY/PROFILE table."
)


print(
    "\n2. The 100-row compatibility table represents "
    "10 workloads × 10 possible servers."
)


print(
    "\n3. The 1-to-1 pairing in the summary table does NOT "
    "force the MILP to assign workload i to server i."
)


print(
    "\n4. UniMelb M_s_raw is NOT yet directly compared "
    "with Bitbrains M_i_req_GB because the UniMelb "
    "RAM physical unit still needs verification."
)
```

```python
# ============================================================
# FINAL 10-ROW THESIS DATASET
#
# PURPOSE:
#
# Create ONE final row per UniMelb server.
#
# 10 UniMelb servers
#       +
# 10 Bitbrains workload profiles
#       =
# 10-row experimental server-workload table
#
# IMPORTANT:
#
# - Time-series observations are summarized.
# - We DO NOT create i × s × t rows.
# - Each server appears exactly once.
# - Each Bitbrains workload appears exactly once.
# - NO workload profiles are reused.

# Result:
# Server 1  -> Workload 1
# Server 2  -> Workload 2
# ...
# Server 9  -> Workload 9
# Server 10 -> Workload 10
#
# IMPORTANT:
# This is an experimental profile pairing.
# It does NOT mean the Bitbrains VM historically
# ran on the paired UniMelb server.
#
# ============================================================


import pandas as pd
import numpy as np
import os


print("=" * 80)
print("CREATING FINAL 10-SERVER + 10-WORKLOAD THESIS DATASET")
print("=" * 80)


# ============================================================
# 1. GET STANDARDIZED DATA FROM PREVIOUS CODE
# ============================================================

if "servers" not in globals():

    raise NameError(
        "\nThe standardized 'servers' dataframe "
        "does not exist.\n"
        "Run the previous standardization code first."
    )


if "workloads" not in globals():

    raise NameError(
        "\nThe standardized 'workloads' dataframe "
        "does not exist.\n"
        "Run the previous standardization code first."
    )


server_data = servers.copy()

workload_data = workloads.copy()


print(
    "\nOriginal server rows:",
    len(server_data)
)

print(
    "Original workload rows:",
    len(workload_data)
)


# ============================================================
# 2. HELPER FUNCTIONS
# ============================================================

def numeric(series):

    return pd.to_numeric(
        series,
        errors="coerce"
    )


def safe_mean(series):

    values = numeric(series)

    if values.notna().sum() == 0:
        return np.nan

    return values.mean()


def safe_median(series):

    values = numeric(series)

    if values.notna().sum() == 0:
        return np.nan

    return values.median()


def safe_max(series):

    values = numeric(series)

    if values.notna().sum() == 0:
        return np.nan

    return values.max()


def safe_min(series):

    values = numeric(series)

    if values.notna().sum() == 0:
        return np.nan

    return values.min()


def safe_p95(series):

    values = numeric(series).dropna()

    if len(values) == 0:
        return np.nan

    return values.quantile(0.95)


# ============================================================
# 3. CREATE ONE ROW PER UNIMELB SERVER
# ============================================================

print("\n" + "=" * 80)
print("SUMMARIZING UNIMELB")
print("=" * 80)


server_rows = []


for server_id, group in server_data.groupby(
    "server_id"
):

    row = {}

    row["server_id"] = int(server_id)


    # --------------------------------------------------------
    # SERVER NAME
    # --------------------------------------------------------

    if "server_name" in group.columns:

        names = (
            group["server_name"]
            .dropna()
            .astype(str)
        )

        row["server_name"] = (
            names.iloc[0]
            if len(names) > 0
            else f"Server_{server_id}"
        )


    # --------------------------------------------------------
    # CPU CAPACITY
    # --------------------------------------------------------

    if "server_CPU_cores" in group.columns:

        row["C_s"] = safe_max(
            group["server_CPU_cores"]
        )


    # --------------------------------------------------------
    # CPU LOAD
    # --------------------------------------------------------

    if "server_CPU_load" in group.columns:

        row["CPU_load_mean"] = safe_mean(
            group["server_CPU_load"]
        )

        row["CPU_load_p95"] = safe_p95(
            group["server_CPU_load"]
        )

        row["CPU_load_max"] = safe_max(
            group["server_CPU_load"]
        )


    # --------------------------------------------------------
    # NORMALIZED CPU LOAD
    # --------------------------------------------------------

    if "server_CPU_load" in group.columns:

        cpu_load = numeric(
            group["server_CPU_load"]
        )

        cpu_min = cpu_load.min()
        cpu_max = cpu_load.max()


        if (
            pd.notna(cpu_min)
            and
            pd.notna(cpu_max)
            and
            cpu_max > cpu_min
        ):

            normalized_cpu = (
                (cpu_load - cpu_min)
                /
                (cpu_max - cpu_min)
            )

            row[
                "server_CPU_load_normalized_mean"
            ] = normalized_cpu.mean()

        else:

            row[
                "server_CPU_load_normalized_mean"
            ] = 0


    # --------------------------------------------------------
    # RAM CAPACITY
    # --------------------------------------------------------

    if "server_RAM" in group.columns:

        row["M_s_raw"] = safe_max(
            group["server_RAM"]
        )


    # --------------------------------------------------------
    # RAM USED
    # --------------------------------------------------------

    if "server_RAM_used" in group.columns:

        row["RAM_used_mean"] = safe_mean(
            group["server_RAM_used"]
        )

        row["RAM_used_p95"] = safe_p95(
            group["server_RAM_used"]
        )


    # --------------------------------------------------------
    # MEMORY UTILIZATION
    # --------------------------------------------------------

    if (
        "server_RAM" in group.columns
        and
        "server_RAM_used" in group.columns
    ):

        ram_total = numeric(
            group["server_RAM"]
        )

        ram_used = numeric(
            group["server_RAM_used"]
        )


        ratio = np.where(
            ram_total > 0,
            ram_used / ram_total,
            np.nan
        )


        ratio = pd.Series(
            ratio,
            index=group.index
        )


        ratio = ratio.where(
            (ratio >= 0)
            &
            (ratio <= 1)
        )


        row[
            "server_memory_utilization_mean"
        ] = ratio.mean()


        row[
            "server_memory_utilization_p95"
        ] = ratio.quantile(0.95)


    # --------------------------------------------------------
    # POWER
    # --------------------------------------------------------

    if "server_power" in group.columns:

        row["P_s_mean"] = safe_mean(
            group["server_power"]
        )

        row["P_s_median"] = safe_median(
            group["server_power"]
        )

        row["P_s_p95"] = safe_p95(
            group["server_power"]
        )

        row["P_s_max"] = safe_max(
            group["server_power"]
        )


    # --------------------------------------------------------
    # INLET TEMPERATURE
    # --------------------------------------------------------

    if (
        "server_inlet_temperature"
        in group.columns
    ):

        row[
            "inlet_temperature_mean"
        ] = safe_mean(
            group[
                "server_inlet_temperature"
            ]
        )


        row[
            "inlet_temperature_p95"
        ] = safe_p95(
            group[
                "server_inlet_temperature"
            ]
        )


    # --------------------------------------------------------
    # CPU TEMPERATURE
    # --------------------------------------------------------

    if (
        "server_CPU_temperature_avg"
        in group.columns
    ):

        row[
            "CPU_temperature_mean"
        ] = safe_mean(
            group[
                "server_CPU_temperature_avg"
            ]
        )


        row[
            "CPU_temperature_p95"
        ] = safe_p95(
            group[
                "server_CPU_temperature_avg"
            ]
        )


        row[
            "CPU_temperature_max"
        ] = safe_max(
            group[
                "server_CPU_temperature_avg"
            ]
        )


    # --------------------------------------------------------
    # FAN SPEED
    # --------------------------------------------------------

    if (
        "server_fan_speed_avg"
        in group.columns
    ):

        row[
            "fan_speed_mean"
        ] = safe_mean(
            group[
                "server_fan_speed_avg"
            ]
        )


    # --------------------------------------------------------
    # NETWORK
    # --------------------------------------------------------

    if (
        "server_network_RX" in group.columns
        and
        "server_network_TX" in group.columns
    ):

        rx = numeric(
            group["server_network_RX"]
        )

        tx = numeric(
            group["server_network_TX"]
        )


        total_network = rx + tx


        row[
            "server_network_mean"
        ] = total_network.mean()


        row[
            "server_network_p95"
        ] = total_network.quantile(
            0.95
        )


    # --------------------------------------------------------
    # NUMBER OF RUNNING VMs
    # --------------------------------------------------------

    if (
        "server_running_VMs"
        in group.columns
    ):

        row[
            "running_VMs_mean"
        ] = safe_mean(
            group[
                "server_running_VMs"
            ]
        )


        row[
            "running_VMs_max"
        ] = safe_max(
            group[
                "server_running_VMs"
            ]
        )


    # --------------------------------------------------------
    # NUMBER OF ORIGINAL OBSERVATIONS
    # --------------------------------------------------------

    row[
        "server_observations"
    ] = len(group)


    server_rows.append(row)


server_summary = pd.DataFrame(
    server_rows
)


server_summary = (
    server_summary
    .sort_values("server_id")
    .reset_index(drop=True)
)


print(
    "\nServer summary rows:",
    len(server_summary)
)


# ============================================================
# 4. CHECK THAT THERE ARE EXACTLY 10 SERVERS
# ============================================================

if len(server_summary) != 10:

    raise ValueError(
        f"\nExpected exactly 10 servers, "
        f"but found {len(server_summary)}."
    )


print(
    "✓ Exactly 10 UniMelb servers."
)


# ============================================================
# 5. CREATE ONE ROW PER BITBRAINS WORKLOAD
# ============================================================

print("\n" + "=" * 80)
print("SUMMARIZING BITBRAINS — 10 WORKLOADS")
print("=" * 80)


workload_rows = []


for workload_id, group in workload_data.groupby(
    "workload_id"
):

    row = {}

    row[
        "workload_id"
    ] = int(workload_id)


    # --------------------------------------------------------
    # ORIGINAL VM ID
    # --------------------------------------------------------

    if (
        "original_VM_ID"
        in group.columns
    ):

        ids = (
            group[
                "original_VM_ID"
            ]
            .dropna()
        )


        if len(ids) > 0:

            row[
                "original_VM_ID"
            ] = ids.iloc[0]


    # --------------------------------------------------------
    # CPU REQUIREMENT
    # --------------------------------------------------------

    if "C_i_req" in group.columns:

        row[
            "C_i_req_mean"
        ] = safe_mean(
            group["C_i_req"]
        )


        row[
            "C_i_req_p95"
        ] = safe_p95(
            group["C_i_req"]
        )


        row[
            "C_i_req_max"
        ] = safe_max(
            group["C_i_req"]
        )


    # --------------------------------------------------------
    # CPU UTILIZATION
    # --------------------------------------------------------

    if (
        "workload_CPU_utilization"
        in group.columns
    ):

        row[
            "workload_CPU_utilization_mean"
        ] = safe_mean(
            group[
                "workload_CPU_utilization"
            ]
        )


        row[
            "workload_CPU_utilization_p95"
        ] = safe_p95(
            group[
                "workload_CPU_utilization"
            ]
        )


    # --------------------------------------------------------
    # CPU USAGE MHz
    # --------------------------------------------------------

    if (
        "workload_CPU_usage_MHz"
        in group.columns
    ):

        row[
            "CPU_usage_MHz_mean"
        ] = safe_mean(
            group[
                "workload_CPU_usage_MHz"
            ]
        )


        row[
            "CPU_usage_MHz_p95"
        ] = safe_p95(
            group[
                "workload_CPU_usage_MHz"
            ]
        )


    # --------------------------------------------------------
    # MEMORY REQUIREMENT GB
    # --------------------------------------------------------

    if (
        "M_i_req_GB"
        in group.columns
    ):

        row[
            "M_i_req_GB_mean"
        ] = safe_mean(
            group[
                "M_i_req_GB"
            ]
        )


        row[
            "M_i_req_GB_p95"
        ] = safe_p95(
            group[
                "M_i_req_GB"
            ]
        )


        row[
            "M_i_req_GB_max"
        ] = safe_max(
            group[
                "M_i_req_GB"
            ]
        )


    # --------------------------------------------------------
    # MEMORY UTILIZATION
    # --------------------------------------------------------

    if (
        "workload_memory_utilization"
        in group.columns
    ):

        row[
            "workload_memory_utilization_mean"
        ] = safe_mean(
            group[
                "workload_memory_utilization"
            ]
        )


        row[
            "workload_memory_utilization_p95"
        ] = safe_p95(
            group[
                "workload_memory_utilization"
            ]
        )


    # --------------------------------------------------------
    # NETWORK
    # --------------------------------------------------------

    if (
        "workload_network_total_KBps"
        in group.columns
    ):

        row[
            "workload_network_KBps_mean"
        ] = safe_mean(
            group[
                "workload_network_total_KBps"
            ]
        )


        row[
            "workload_network_KBps_p95"
        ] = safe_p95(
            group[
                "workload_network_total_KBps"
            ]
        )


    # --------------------------------------------------------
    # DISK READ
    # --------------------------------------------------------

    if (
        "workload_disk_read_KBps"
        in group.columns
    ):

        row[
            "disk_read_KBps_mean"
        ] = safe_mean(
            group[
                "workload_disk_read_KBps"
            ]
        )


    # --------------------------------------------------------
    # DISK WRITE
    # --------------------------------------------------------

    if (
        "workload_disk_write_KBps"
        in group.columns
    ):

        row[
            "disk_write_KBps_mean"
        ] = safe_mean(
            group[
                "workload_disk_write_KBps"
            ]
        )


    # --------------------------------------------------------
    # NUMBER OF ORIGINAL OBSERVATIONS
    # --------------------------------------------------------

    row[
        "workload_observations"
    ] = len(group)


    workload_rows.append(row)


workload_summary = pd.DataFrame(
    workload_rows
)


workload_summary = (
    workload_summary
    .sort_values("workload_id")
    .reset_index(drop=True)
)


print(
    "\nWorkload summary rows:",
    len(workload_summary)
)


# ============================================================
# 6. CHECK FOR EXACTLY 10 WORKLOADS
# ============================================================

if len(workload_summary) != 10:

    raise ValueError(
        f"\nExpected exactly 10 Bitbrains workloads, "
        f"but found {len(workload_summary)}.\n\n"
        "Make sure the previous Bitbrains preparation code "
        "loaded files 1.csv through 10.csv."
    )


if workload_summary[
    "workload_id"
].nunique() != 10:

    raise ValueError(
        "\nExpected 10 unique workload IDs."
    )


print(
    "✓ Exactly 10 Bitbrains workloads."
)


print(
    "Workload IDs:",
    workload_summary[
        "workload_id"
    ].tolist()
)


# ============================================================
# 7. PAIR 10 WORKLOAD PROFILES WITH 10 SERVERS
# ============================================================
#
# NO CYCLIC REUSE IS NEEDED.
#
# Server 1  -> Workload 1
# Server 2  -> Workload 2
# ...
# Server 10 -> Workload 10
#
# This is ONLY an experimental summary pairing.
# It does NOT constrain the later MILP assignment X_is.
# ============================================================


server_summary = (
    server_summary
    .sort_values("server_id")
    .reset_index(drop=True)
)


workload_summary = (
    workload_summary
    .sort_values("workload_id")
    .reset_index(drop=True)
)


server_summary[
    "assigned_workload_id"
] = workload_summary[
    "workload_id"
].to_numpy()


print(
    "\nExperimental workload assignment:"
)


print(
    server_summary[
        [
            "server_id",
            "server_name",
            "assigned_workload_id"
        ]
    ].to_string(
        index=False
    )
)


# ============================================================
# 8. MERGE WORKLOAD PROFILE INTO EACH SERVER
# ============================================================

workload_for_merge = (
    workload_summary
    .rename(
        columns={
            "workload_id":
                "assigned_workload_id"
        }
    )
)


final_10_rows = (
    server_summary

    .merge(
        workload_for_merge,

        on="assigned_workload_id",

        how="left",

        validate="one_to_one"
    )
)


# ============================================================
# 9. ADD REUSE FLAG
# ============================================================
#
# Because there are 10 unique workloads for 10 servers,
# no workload should be reused.
# ============================================================

final_10_rows[
    "workload_profile_reused"
] = False


# ============================================================
# 10. ADD DATASET ROLE LABELS
# ============================================================

final_10_rows[
    "server_data_source"
] = "UniMelb"


final_10_rows[
    "workload_data_source"
] = "Bitbrains"


final_10_rows[
    "pairing_type"
] = (
    "Experimental simulation pairing"
)


# ============================================================
# 11. REORDER IMPORTANT COLUMNS
# ============================================================

front_columns = [

    "server_id",

    "server_name",

    "assigned_workload_id",

    "original_VM_ID",

    "server_data_source",

    "workload_data_source",

    "pairing_type",

    "workload_profile_reused",

    "C_s",

    "C_i_req_mean",

    "C_i_req_p95",

    "C_i_req_max",

    "M_s_raw",

    "M_i_req_GB_mean",

    "M_i_req_GB_p95",

    "M_i_req_GB_max",

    "CPU_load_mean",

    "server_CPU_load_normalized_mean",

    "workload_CPU_utilization_mean",

    "server_memory_utilization_mean",

    "workload_memory_utilization_mean",

    "P_s_mean",

    "CPU_temperature_mean",

    "inlet_temperature_mean"
]


front_columns = [
    c
    for c in front_columns
    if c in final_10_rows.columns
]


remaining_columns = [
    c
    for c in final_10_rows.columns
    if c not in front_columns
]


final_10_rows = final_10_rows[
    front_columns
    +
    remaining_columns
]


# ============================================================
# 12. FINAL QUALITY CHECK
# ============================================================

print("\n" + "=" * 80)
print("FINAL DATASET CHECK")
print("=" * 80)


print(
    "\nFinal number of rows:",
    len(final_10_rows)
)


print(
    "Unique servers:",
    final_10_rows[
        "server_id"
    ].nunique()
)


print(
    "Unique workload profiles:",
    final_10_rows[
        "assigned_workload_id"
    ].nunique()
)


if "original_VM_ID" in final_10_rows.columns:

    print(
        "Unique original Bitbrains VMs:",
        final_10_rows[
            "original_VM_ID"
        ].nunique()
    )


print(
    "\nExpected:"
)

print(
    "10 rows"
)

print(
    "10 unique servers"
)

print(
    "10 unique Bitbrains workloads"
)


if (
    len(final_10_rows) == 10

    and

    final_10_rows[
        "server_id"
    ].nunique() == 10

    and

    final_10_rows[
        "assigned_workload_id"
    ].nunique() == 10
):

    print(
        "\n✓ SUCCESS:"
    )

    print(
        "Final table contains exactly "
        "10 servers + 10 unique workloads."
    )

else:

    raise ValueError(
        "\nFinal table failed the "
        "10-server / 10-workload check."
    )


# ============================================================
# 13. SAVE FINAL TABLE
# ============================================================

OUTPUT_FOLDER = (
    "/content/THESIS_FINAL_10_ROWS"
)


os.makedirs(
    OUTPUT_FOLDER,
    exist_ok=True
)


FINAL_FILE = os.path.join(
    OUTPUT_FOLDER,
    "FINAL_10_SERVER_10_WORKLOAD_DATASET.csv"
)


final_10_rows.to_csv(
    FINAL_FILE,
    index=False
)


# ============================================================
# 14. SAVE SERVER AND WORKLOAD SUMMARIES
# ============================================================

SERVER_SUMMARY_FILE = os.path.join(
    OUTPUT_FOLDER,
    "UNIMELB_10_SERVER_SUMMARY.csv"
)


WORKLOAD_SUMMARY_FILE = os.path.join(
    OUTPUT_FOLDER,
    "BITBRAINS_10_WORKLOAD_SUMMARY.csv"
)


server_summary.to_csv(
    SERVER_SUMMARY_FILE,
    index=False
)


workload_summary.to_csv(
    WORKLOAD_SUMMARY_FILE,
    index=False
)


# ============================================================
# 15. DISPLAY FINAL TABLE
# ============================================================

print("\n" + "=" * 80)
print("FINAL 10-ROW TABLE")
print("=" * 80)


display(
    final_10_rows
)


# ============================================================
# 16. DISPLAY SIMPLE SERVER-WORKLOAD MAPPING
# ============================================================

print("\n" + "=" * 80)
print("SERVER-WORKLOAD MAPPING")
print("=" * 80)


mapping_columns = [

    "server_id",

    "server_name",

    "assigned_workload_id",

    "original_VM_ID"
]


mapping_columns = [
    c
    for c in mapping_columns
    if c in final_10_rows.columns
]


display(
    final_10_rows[
        mapping_columns
    ]
)


# ============================================================
# 17. FILE LOCATIONS
# ============================================================

print("\n" + "=" * 80)
print("FILES CREATED")
print("=" * 80)


print(
    "\nFINAL 10-row dataset:"
)

print(
    FINAL_FILE
)


print(
    "\n10-server UniMelb summary:"
)

print(
    SERVER_SUMMARY_FILE
)


print(
    "\n10-workload Bitbrains summary:"
)

print(
    WORKLOAD_SUMMARY_FILE
)
```

```python
import pandas as pd
import numpy as np

# ============================================================
# 1. LOAD THE 10-SERVER UNIMELB DATASET
# ============================================================

file_path = "/content/dataset1_10servers.csv"

df = pd.read_csv(file_path)

# Clean column names
df.columns = df.columns.str.strip()

print("Columns found:")
print(df.columns.tolist())


# ============================================================
# 2. MAKE SURE CPU LOAD AND POWER ARE NUMERIC
# ============================================================

df["CPU_Load"] = pd.to_numeric(df["CPU_Load"], errors="coerce")
df["Power"] = pd.to_numeric(df["Power"], errors="coerce")

# Remove rows where CPU load or power is missing
df_clean = df.dropna(subset=["Series", "CPU_Load", "Power"]).copy()


# ============================================================
# 3. FIND MINIMUM CPU LOAD FOR EACH SERVER
# ============================================================

min_cpu = (
    df_clean
    .groupby("Series")["CPU_Load"]
    .min()
    .reset_index()
    .rename(columns={"CPU_Load": "Min_CPU_Load"})
)


# ============================================================
# 4. GET OBSERVATIONS AT EACH SERVER'S MINIMUM CPU LOAD
# ============================================================

min_load_rows = df_clean.merge(
    min_cpu,
    on="Series",
    how="left"
)

# Keep rows where CPU load equals that server's minimum load
min_load_rows = min_load_rows[
    np.isclose(
        min_load_rows["CPU_Load"],
        min_load_rows["Min_CPU_Load"]
    )
].copy()


# ============================================================
# 5. CALCULATE ESTIMATED IDLE POWER
# ============================================================

idle_power = (
    min_load_rows
    .groupby("Series")
    .agg(
        Min_CPU_Load=("Min_CPU_Load", "first"),
        Number_of_Min_Load_Observations=("Power", "count"),
        P_idle=("Power", "mean")
    )
    .reset_index()
)


# ============================================================
# 6. ADD CHECK: IS THE MINIMUM ACTUALLY 0%?
# ============================================================

idle_power["Exact_Zero_CPU"] = np.isclose(
    idle_power["Min_CPU_Load"], 0
)

idle_power["Method_Note"] = np.where(
    idle_power["Exact_Zero_CPU"],
    "Exact 0% CPU load observed",
    "No exact 0%; uses minimum observed CPU load"
)


# ============================================================
# 7. CALCULATE PMAX USING HSIEH ET AL. (2020)
#
# P_idle = 0.7 P_max
#
# Therefore:
# P_max = P_idle / 0.7
# ============================================================

idle_power["P_max"] = idle_power["P_idle"] / 0.70


# ============================================================
# 8. DISPLAY RESULTS
# ============================================================

print("\n" + "=" * 80)
print("ESTIMATED IDLE AND MAXIMUM POWER PER SERVER")
print("=" * 80)

print(idle_power.to_string(index=False))


# ============================================================
# 9. SAVE RESULTS
# ============================================================

output_path = "/content/unimelb_idle_max_power.csv"

idle_power.to_csv(output_path, index=False)

print("\nSaved to:")
print(output_path)
```

```python
import pandas as pd
import numpy as np

# ============================================================
# ESTIMATE IDLE POWER USING CPU LOAD <= 1%
# ============================================================

file_path = "/content/dataset1_10servers.csv"

df = pd.read_csv(file_path)
df.columns = df.columns.str.strip()

df["CPU_Load"] = pd.to_numeric(df["CPU_Load"], errors="coerce")
df["Power"] = pd.to_numeric(df["Power"], errors="coerce")

df = df.dropna(subset=["Series", "CPU_Load", "Power"]).copy()


# ------------------------------------------------------------
# 1. Select near-idle observations
#    CPU_Load is already expressed in percentage points.
#    Therefore 1.0 means 1% CPU utilization.
# ------------------------------------------------------------

near_idle = df[df["CPU_Load"] <= 1.0].copy()


# ------------------------------------------------------------
# 2. Calculate idle-power statistics for each server
# ------------------------------------------------------------

idle_summary = (
    near_idle
    .groupby("Series")
    .agg(
        Near_Idle_Observations=("Power", "count"),
        Min_CPU_Load=("CPU_Load", "min"),
        Max_CPU_Load_In_Idle_Group=("CPU_Load", "max"),
        Mean_CPU_Load_In_Idle_Group=("CPU_Load", "mean"),
        P_idle=("Power", "mean"),
        P_idle_std=("Power", "std"),
        P_idle_min=("Power", "min"),
        P_idle_max=("Power", "max")
    )
    .reset_index()
)


# ------------------------------------------------------------
# 3. Calculate Pmax using:
#
#       Pidle = 0.7 Pmax
#
#    therefore:
#
#       Pmax = Pidle / 0.7
# ------------------------------------------------------------

idle_summary["P_max"] = idle_summary["P_idle"] / 0.70


# ------------------------------------------------------------
# 4. Display
# ------------------------------------------------------------

print("=" * 110)
print("NEAR-IDLE POWER ESTIMATION (CPU LOAD <= 1%)")
print("=" * 110)

print(idle_summary.to_string(index=False))


# ------------------------------------------------------------
# 5. Save
# ------------------------------------------------------------

output_path = "/content/unimelb_idle_power_1percent.csv"

idle_summary.to_csv(output_path, index=False)

print("\nSaved to:")
print(output_path)
```

```python
import pandas as pd

# ============================================================
# LOAD DATA
# ============================================================

file_path = "/content/dataset1_10servers.csv"

df = pd.read_csv(file_path)
df.columns = df.columns.str.strip()

df["CPU_Load"] = pd.to_numeric(df["CPU_Load"], errors="coerce")
df["Power"] = pd.to_numeric(df["Power"], errors="coerce")

df = df.dropna(subset=["Series", "CPU_Load", "Power"]).copy()


# ============================================================
# 1. ESTIMATE PIDLE
#    Average power when CPU load <= 1%
# ============================================================

near_idle = df[df["CPU_Load"] <= 1.0].copy()

p_idle = (
    near_idle
    .groupby("Series")["Power"]
    .mean()
    .reset_index(name="P_idle")
)


# ============================================================
# 2. GET HIGHEST OBSERVED POWER AND CPU LOAD
# ============================================================

server_observed = (
    df.groupby("Series")
    .agg(
        Highest_Observed_Power=("Power", "max"),
        Max_CPU_Load=("CPU_Load", "max")
    )
    .reset_index()
)


# ============================================================
# 3. COMBINE RESULTS
# ============================================================

check = p_idle.merge(
    server_observed,
    on="Series"
)


# ============================================================
# 4. CALCULATE OBSERVED IDLE/HIGHEST-POWER RATIO
# ============================================================

check["Observed_Ratio"] = (
    check["P_idle"] /
    check["Highest_Observed_Power"]
)

check["Observed_Ratio_Percent"] = (
    check["Observed_Ratio"] * 100
)


# ============================================================
# 5. HSIEH-BASED PMAX
#
# Pidle = 0.70 Pmax
# Pmax = Pidle / 0.70
# ============================================================

check["Hsieh_Estimated_Pmax"] = (
    check["P_idle"] / 0.70
)


# ============================================================
# DISPLAY RESULTS
# ============================================================

print("=" * 100)
print("CHECK OF HSIEH 70% ASSUMPTION AGAINST UNIMELB DATA")
print("=" * 100)

print(
    check[
        [
            "Series",
            "P_idle",
            "Highest_Observed_Power",
            "Max_CPU_Load",
            "Observed_Ratio_Percent",
            "Hsieh_Estimated_Pmax"
        ]
    ].to_string(index=False)
)
```

```python
# ============================================================
# TEST Pmax USING CPU LOAD vs POWER REGRESSION
# UniMelb - 10 Selected Servers
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import os

# ------------------------------------------------------------
# 1. FILE PATH
# ------------------------------------------------------------

# Change this only if your file has a different name/path
FILE_PATH = "/content/dataset1_10servers.csv"

# Output files
OUTPUT_PATH = "/content/unimelb_pmax_regression.csv"
GRAPH_FOLDER = "/content/pmax_regression_graphs"

os.makedirs(GRAPH_FOLDER, exist_ok=True)


# ------------------------------------------------------------
# 2. LOAD DATA
# ------------------------------------------------------------

df = pd.read_csv(FILE_PATH)

# Clean column names
df.columns = df.columns.str.strip()

print("=" * 80)
print("COLUMNS FOUND")
print("=" * 80)
print(df.columns.tolist())


# ------------------------------------------------------------
# 3. CHECK REQUIRED COLUMNS
# ------------------------------------------------------------

required_columns = ["Series", "CPU_Load", "Power"]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )


# ------------------------------------------------------------
# 4. CLEAN CPU LOAD AND POWER
# ------------------------------------------------------------

df["CPU_Load"] = pd.to_numeric(
    df["CPU_Load"],
    errors="coerce"
)

df["Power"] = pd.to_numeric(
    df["Power"],
    errors="coerce"
)

# Remove rows where CPU load or power is missing
df = df.dropna(
    subset=["Series", "CPU_Load", "Power"]
).copy()


# ------------------------------------------------------------
# 5. NORMALIZE CPU LOAD
# ------------------------------------------------------------

# IMPORTANT:
# UniMelb CPU_Load is in percentage points.
#
# Example:
# 1.0  = 1% CPU
# 25.0 = 25% CPU
# 100  = 100% CPU
#
# Convert it to 0-1 for the mathematical model.

df["CPU_Utilization"] = df["CPU_Load"] / 100.0


# ------------------------------------------------------------
# 6. RUN REGRESSION FOR EACH SERVER
# ------------------------------------------------------------

results = []

servers = sorted(df["Series"].unique())

print()
print("=" * 80)
print("REGRESSION RESULTS")
print("=" * 80)

for server in servers:

    server_df = df[
        df["Series"] == server
    ].copy()

    # X = CPU utilization
    X = server_df[["CPU_Utilization"]]

    # y = observed server power
    y = server_df["Power"]

    # Fit linear regression
    model = LinearRegression()
    model.fit(X, y)

    # --------------------------------------------------------
    # Regression equation:
    #
    # P_obs = alpha + beta * U
    #
    # alpha = predicted power at 0% CPU
    # beta  = increase in power from U = 0 to U = 1
    # --------------------------------------------------------

    alpha = model.intercept_
    beta = model.coef_[0]

    # Predicted values for R²
    y_pred = model.predict(X)

    r2 = r2_score(y, y_pred)

    # --------------------------------------------------------
    # Estimate Pmax at 100% CPU utilization
    #
    # At U = 1:
    #
    # Pmax = alpha + beta
    # --------------------------------------------------------

    Pmax_estimated = alpha + beta

    # Actual maximums observed in the dataset
    max_cpu_load = server_df["CPU_Load"].max()
    max_observed_power = server_df["Power"].max()

    # Minimums for reference
    min_cpu_load = server_df["CPU_Load"].min()
    min_observed_power = server_df["Power"].min()

    # Number of observations
    n = len(server_df)

    results.append({
        "Series": server,
        "Observations": n,

        "Alpha_Intercept_W": alpha,
        "Beta_Slope_W": beta,

        "Estimated_Pmax_at_100_CPU_W": Pmax_estimated,

        "R_squared": r2,

        "Min_CPU_Load_percent": min_cpu_load,
        "Max_CPU_Load_percent": max_cpu_load,

        "Min_Observed_Power_W": min_observed_power,
        "Max_Observed_Power_W": max_observed_power
    })


    # --------------------------------------------------------
    # 7. CREATE GRAPH FOR EACH SERVER
    # --------------------------------------------------------

    plt.figure(figsize=(8, 5))

    # Actual observations
    plt.scatter(
        server_df["CPU_Load"],
        server_df["Power"],
        s=10,
        alpha=0.25,
        label="Observed UniMelb data"
    )

    # Regression line from 0% to 100%
    cpu_range_percent = np.linspace(0, 100, 200)

    cpu_range_normalized = (
        cpu_range_percent / 100
    ).reshape(-1, 1)

    predicted_power = model.predict(
        cpu_range_normalized
    )

    plt.plot(
        cpu_range_percent,
        predicted_power,
        linewidth=2,
        label="Linear regression"
    )

    # Show where the real dataset ends
    plt.axvline(
        max_cpu_load,
        linestyle="--",
        label=f"Highest observed CPU = {max_cpu_load:.1f}%"
    )

    # Show estimated Pmax
    plt.scatter(
        [100],
        [Pmax_estimated],
        s=80,
        label=f"Estimated Pmax = {Pmax_estimated:.1f} W"
    )

    plt.xlabel("CPU Load (%)")
    plt.ylabel("Power (W)")

    plt.title(
        f"{server}\n"
        f"CPU Load vs Power | "
        f"R² = {r2:.3f} | "
        f"Estimated Pmax = {Pmax_estimated:.1f} W"
    )

    plt.legend()
    plt.grid(alpha=0.2)
    plt.tight_layout()

    graph_path = os.path.join(
        GRAPH_FOLDER,
        f"{server}_CPU_vs_Power.png"
    )

    plt.savefig(
        graph_path,
        dpi=200,
        bbox_inches="tight"
    )

    plt.close()


# ------------------------------------------------------------
# 8. CREATE RESULTS TABLE
# ------------------------------------------------------------

results_df = pd.DataFrame(results)

# Round values for easier reading
numeric_columns = results_df.select_dtypes(
    include=[np.number]
).columns

results_df[numeric_columns] = (
    results_df[numeric_columns].round(4)
)


# ------------------------------------------------------------
# 9. PRINT RESULTS
# ------------------------------------------------------------

print(results_df.to_string(index=False))


# ------------------------------------------------------------
# 10. SAVE RESULTS
# ------------------------------------------------------------

results_df.to_csv(
    OUTPUT_PATH,
    index=False
)

print()
print("=" * 80)
print("FILES SAVED")
print("=" * 80)

print("Regression results:")
print(OUTPUT_PATH)

print()
print("Graphs:")
print(GRAPH_FOLDER)


# ------------------------------------------------------------
# 11. SIMPLE R² INTERPRETATION
# ------------------------------------------------------------

print()
print("=" * 80)
print("R² CHECK")
print("=" * 80)

for _, row in results_df.iterrows():

    r2 = row["R_squared"]

    # These labels are only simple diagnostics,
    # not literature-based acceptance thresholds.
    if r2 >= 0.70:
        message = "Strong linear relationship"

    elif r2 >= 0.40:
        message = "Moderate linear relationship"

    else:
        message = "Weak linear relationship - inspect carefully"

    print(
        f"{row['Series']}: "
        f"R² = {r2:.3f} -> {message}"
    )
```
